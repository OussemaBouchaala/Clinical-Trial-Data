```python
# PURPOSE: Understand how pycountry works before using it.
# WHY: Like testing a new tool on one screw before using it on the whole shelf.
# pycountry is a library that has a list of all countries in the world.
# We give it a messy name → it gives back the official name.

import pycountry

# Test with a few examples
test_names = ["United States", "USA", "US", "Turkey (Türkiye)", "Korea, Republic of", "Brazil"]

for name in test_names:
    # Try exact match first
    result = pycountry.countries.get(name=name)
    if not result:
        # Try fuzzy search if exact fails
        try:
            result = pycountry.countries.search_fuzzy(name)[0]
        except LookupError:
            result = None
    
    clean = result.name if result else "NOT FOUND"
    print(f"  '{name}' → '{clean}'")
```

```python
# PURPOSE: Create a function that cleans ANY country name.
# WHY: pycountry works for most names, but some need manual help.
# Strategy: 1) Check manual fixes first → 2) Try pycountry exact → 3) Try fuzzy search
# Like a 3-step filter: easy fixes first, then smart lookup, then best guess.

import pycountry

# Manual fixes for names pycountry can't handle
COUNTRY_MANUAL_FIXES = {
    "turkey (türkiye)": "Turkey",
    "türkiye": "Turkey",
    "brasil": "Brazil",
    "korea, republic of": "South Korea",
    "republic of korea": "South Korea",
    "korea, democratic people's republic of": "North Korea",
    "russian federation": "Russia",
    "iran, islamic republic of": "Iran",
    "taiwan, province of china": "Taiwan",
    "viet nam": "Vietnam",
    "congo, the democratic republic of the": "Democratic Republic of the Congo",
    "tanzania, united republic of": "Tanzania",
    "palestinian territories": "Palestine",
    "faroe islands": "Faroe Islands",
}

def clean_country(name):
    """Takes a messy country name, returns a clean one."""
    if pd.isna(name):
        return name
    
    name_lower = name.strip().lower()
    
    # Step 1: Check manual fixes
    if name_lower in COUNTRY_MANUAL_FIXES:
        return COUNTRY_MANUAL_FIXES[name_lower]
    
    # Step 2: Try pycountry exact match
    result = pycountry.countries.get(name=name.strip())
    if result:
        return result.name
    
    # Step 3: Try fuzzy search
    try:
        result = pycountry.countries.search_fuzzy(name.strip())[0]
        return result.name
    except LookupError:
        return name.strip()  # give up, return as-is but trimmed

# Test it
test = ["Turkey (Türkiye)", "Brasil", "USA", "United States", "Korea, Republic of"]
for t in test:
    print(f"  '{t}' → '{clean_country(t)}'")
```

```python
# PURPOSE: Clean every country in the dataset.
# WHY: Remember, some rows have MULTIPLE countries like "France, Germany, Italy".
# So we: 1) Split by comma → 2) Clean each one → 3) Join back together.
# Like washing each sock individually then putting them back in the same drawer.

def clean_country_cell(cell):
    """Handles cells with multiple comma-separated countries."""
    if pd.isna(cell):
        return cell
    
    # Split by comma, clean each, rejoin
    countries = [clean_country(c.strip()) for c in cell.split(",")]
    # Remove duplicates while keeping order
    seen = set()
    unique = []
    for c in countries:
        if c not in seen:
            seen.add(c)
            unique.append(c)
    
    return ", ".join(unique)

# Apply to the whole column
df['country_cleaned'] = df['country'].apply(clean_country_cell)

# Check results
print(f"Before: {df['country'].nunique()} unique values")
print(f"After:  {df['country_cleaned'].nunique()} unique values")
```

```python
# PURPOSE: Some country names CONTAIN commas like "Korea, Republic of"
# or "Tanzania, United Republic of". Our comma-split broke them apart.
# FIX: Replace these tricky names BEFORE splitting by comma.
# Like replacing trick words before cutting a sentence into pieces.

# These are names that have commas inside them — we swap them first
COMMA_NAMES_FIX = {
    "Korea, Republic of": "South Korea",
    "Korea, Democratic People's Republic of": "North Korea",
    "Congo, The Democratic Republic of the": "Democratic Republic of the Congo",
    "Congo, the Democratic Republic of the": "Democratic Republic of the Congo",
    "Tanzania, United Republic of": "Tanzania",
    "Iran, Islamic Republic of": "Iran",
    "Taiwan, Province of China": "Taiwan",
    "Venezuela, Bolivarian Republic of": "Venezuela",
    "Bolivia, Plurinational State of": "Bolivia",
    "Moldova, Republic of": "Moldova",
    "Micronesia, Federated States of": "Micronesia",
    "Palestine, State of": "Palestine",
    "Lao People's Democratic Republic": "Laos",
    "Syrian Arab Republic": "Syria",
    "Viet Nam": "Vietnam",
    "Russian Federation": "Russia",
    "Czechia": "Czech Republic",
    "Côte d'Ivoire": "Ivory Coast",
    "Réunion": "Reunion",
}

def clean_country_cell_v2(cell):
    """Improved version: fixes comma-names BEFORE splitting."""
    if pd.isna(cell):
        return cell
    
    text = cell.strip()
    
    # Step 1: Replace known comma-containing names before splitting
    for messy, clean in COMMA_NAMES_FIX.items():
        text = text.replace(messy, clean)
    
    # Step 2: Now safe to split by comma
    countries = [clean_country(c.strip()) for c in text.split(",")]
    
    # Step 3: Remove duplicates keeping order
    seen = set()
    unique = []
    for c in countries:
        if c not in seen:
            seen.add(c)
            unique.append(c)
    
    return ", ".join(unique)

# Re-apply
df['country_cleaned'] = df['country'].apply(clean_country_cell_v2)

print(f"Before: {df['country'].nunique()} unique values")
print(f"After:  {df['country_cleaned'].nunique()} unique values")

# Check individual countries again
all_countries = df['country_cleaned'].dropna().str.split(", ").explode().str.strip()
print(f"\nUnique individual countries: {all_countries.nunique()}")
print(f"\nAll values:\n{sorted(all_countries.unique())}")
```

```python
# PURPOSE: Rebuild clean_country() with 3 layers:
# 1) PRE-FIX: handle known messy inputs before pycountry
# 2) PYCOUNTRY: try to match
# 3) POST-FIX: convert ugly official names to common names
# Like: wash → dry → iron. Three steps to clean clothes properly.

# POST-FIX: pycountry's ugly official names → common names
COUNTRY_POST_FIX = {
    "Iran, Islamic Republic of": "Iran",
    "Viet Nam": "Vietnam",
    "Russian Federation": "Russia",
    "Taiwan, Province of China": "Taiwan",
    "Tanzania, United Republic of": "Tanzania",
    "Syrian Arab Republic": "Syria",
    "Lao People's Democratic Republic": "Laos",
    "Venezuela, Bolivarian Republic of": "Venezuela",
    "Moldova, Republic of": "Moldova",
    "Réunion": "Reunion",
    "Korea, Republic of": "South Korea",
    "Korea, Democratic People's Republic of": "North Korea",
    "Congo, The Democratic Republic of the": "Democratic Republic of the Congo",
    "Bolivia, Plurinational State of": "Bolivia",
    "Micronesia, Federated States of": "Micronesia",
    "Palestine, State of": "Palestine",
    "Côte d'Ivoire": "Ivory Coast",
    "Czechia": "Czech Republic",
}

# PRE-FIX: handle names that pycountry can't find at all
COUNTRY_PRE_FIX = {
    "turkey (türkiye)": "Turkey",
    "türkiye": "Turkey",
    "palestinian territories": "Palestine",
    "faroe islands": "Faroe Islands",
    "brasil": "Brazil"
}

def clean_country_v3(name):
    """Final version: pre-fix → pycountry → post-fix."""
    if pd.isna(name):
        return name
    
    trimmed = name.strip()
    lower = trimmed.lower()
    
    # Step 1: Pre-fix
    if lower in COUNTRY_PRE_FIX:
        return COUNTRY_PRE_FIX[lower]
    
    # Step 2: Try pycountry
    result = pycountry.countries.get(name=trimmed)
    if not result:
        try:
            result = pycountry.countries.search_fuzzy(trimmed)[0]
        except LookupError:
            return trimmed
    
    official = result.name
    
    # Step 3: Post-fix
    if official in COUNTRY_POST_FIX:
        return COUNTRY_POST_FIX[official]
    
    return official

# Test
test = ["Iran", "Vietnam", "Russia", "Taiwan", "Tanzania",
        "Turkey (Türkiye)", "USA", "United States", "Brasil"]
for t in test:
    print(f"  '{t}' → '{clean_country_v3(t)}'")
```

```python
# PURPOSE: Clean sponsor names by stripping common suffixes and normalizing.
# WHY: "Novartis Pharmaceuticals" and "Novartis" are the same company.
# Strategy: 
#   1) Strip suffixes like Inc., LLC, Ltd, Co., Pharmaceuticals, etc.
#   2) Normalize whitespace and punctuation
#   3) Use fuzzy matching for near-duplicates
# Like peeling labels off jars to see they're the same brand.

import re

# Suffixes to strip (order matters — longer first)
SUFFIXES_TO_STRIP = [
    r",?\s*Inc\.?",
    r",?\s*LLC\.?",
    r",?\s*Ltd\.?",
    r",?\s*Co\.?",
    r",?\s*Corp\.?",
    r",?\s*Corporation",
    r",?\s*Incorporated",
    r",?\s*Limited",
    r",?\s*S\.?A\.?",
    r",?\s*S\.?p\.?A\.?",
    r",?\s*GmbH",
    r",?\s*AG",
    r",?\s*PLC",
    r",?\s*Pty",
    r",?\s*B\.?V\.?",
    r",?\s*N\.?V\.?",
    r",?\s*SE",
]

def clean_sponsor(name):
    """Clean a single sponsor name."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # Remove suffixes from the END of the name
    for suffix in SUFFIXES_TO_STRIP:
        cleaned = re.sub(suffix + r'\s*$', '', cleaned, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation
    cleaned = cleaned.rstrip('.,;: ')
    
    return cleaned

# Test
tests = ["Pfizer", "Pfizer, Inc.", "Pfizer Inc", 
         "Novartis Pharmaceuticals", "Novartis",
         "Click Therapeutics, Inc.", "Pharmavite LLC",
         "Hoffmann-La Roche", "Eli Lilly and Company"]
for t in tests:
    print(f"  '{t}' → '{clean_sponsor(t)}'")
```

```python
# PURPOSE: Extend suffix stripping to catch "Pharmaceuticals", "Vaccines", 
# "and Company" etc. These are department names, not different companies.
# Like: "John's Bakery Department" → just "John's Bakery" → just "John"

SUFFIXES_TO_STRIP_V2 = [
    # Corporate suffixes
    r",?\s*Inc\.?",
    r",?\s*LLC\.?",
    r",?\s*Ltd\.?",
    r",?\s*Co\.?",
    r",?\s*Corp\.?",
    r",?\s*Corporation",
    r",?\s*Incorporated",
    r",?\s*Limited",
    r",?\s*S\.?A\.?",
    r",?\s*S\.?p\.?A\.?",
    r",?\s*GmbH",
    r",?\s*PLC",
    r",?\s*Pty",
    r",?\s*B\.?V\.?",
    r",?\s*N\.?V\.?",
    r",?\s*SE",
    # Division/department suffixes
    r",?\s*Pharmaceuticals?",
    r",?\s*Vaccines?",
    r",?\s*and Company",
    r",?\s*& Company",
    r",?\s*Biosciences?",
    r",?\s*Therapeutics?",
    r",?\s*Sciences?",
    r",?\s*Biotech",
    r",?\s*Pharma",
]

def clean_sponsor_v2(name):
    """Clean sponsor: strip suffixes, normalize whitespace."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # Run suffix stripping TWICE (for "Novartis Pharmaceuticals, Inc.")
    for _ in range(2):
        for suffix in SUFFIXES_TO_STRIP_V2:
            cleaned = re.sub(suffix + r'\s*$', '', cleaned, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation
    cleaned = cleaned.rstrip('.,;: ')
    
    return cleaned

# Test
tests = ["Novartis Pharmaceuticals", "Novartis", "Novartis Vaccines",
         "Eli Lilly and Company", "Hoffmann-La Roche",
         "Click Therapeutics, Inc.", "Merck Sharp & Dohme LLC",
         "Boehringer Ingelheim", "AstraZeneca"]
for t in tests:
    print(f"  '{t}' → '{clean_sponsor_v2(t)}'")
```

```python
# PURPOSE: Fix over-stripping. "Click Therapeutics" is the real name,
# but "Novartis Pharmaceuticals" is Novartis + department label.
# NEW STRATEGY: 
#   1) Strip corporate suffixes (Inc, LLC, Ltd, etc.) — always safe
#   2) Strip division words (Pharmaceuticals, Vaccines) ONLY for known big pharma
# Like: remove "Mr." from everyone, but only remove "Junior" from known families.

import re

CORPORATE_SUFFIXES = [
    r",?\s*Inc\.?",
    r",?\s*LLC\.?",
    r",?\s*Ltd\.?",
    r",?\s*Co\.?\b",
    r",?\s*Corp\.?",
    r",?\s*Corporation",
    r",?\s*Incorporated",
    r",?\s*Limited",
    r",?\s*S\.?A\.?\b",
    r",?\s*S\.?p\.?A\.?",
    r",?\s*GmbH",
    r",?\s*PLC",
    r",?\s*Pty",
    r",?\s*B\.?V\.?\b",
    r",?\s*N\.?V\.?\b",
    r",?\s*SE\b",
    r",?\s*and Company",
    r",?\s*& Company",
]

# Known big pharma where "Pharmaceuticals"/"Vaccines" is a division label
KNOWN_PHARMA_BASES = [
    "novartis", "pfizer", "roche", "merck", "sanofi", "bayer",
    "johnson", "abbvie", "amgen", "gilead", "bristol-myers",
    "astrazeneca", "glaxosmithkline", "gsk", "takeda", "allergan",
]

def clean_sponsor_v3(name):
    """Clean sponsor with safe suffix stripping."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # Step 1: Strip corporate suffixes (always safe)
    for suffix in CORPORATE_SUFFIXES:
        cleaned = re.sub(suffix + r'\s*$', '', cleaned, flags=re.IGNORECASE).strip()
    
    # Step 2: Strip division words ONLY if base is a known big pharma
    base_lower = cleaned.lower().split()[0] if cleaned else ""
    if any(pharma in cleaned.lower() for pharma in KNOWN_PHARMA_BASES):
        for div in [r"\s+Pharmaceuticals?$", r"\s+Vaccines?$", r"\s+Pharma$", 
                     r"\s+Biosciences?$", r"\s+Biotech$"]:
            cleaned = re.sub(div, '', cleaned, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation
    cleaned = cleaned.rstrip('.,;: ')
    
    return cleaned

# Test
tests = ["Novartis Pharmaceuticals", "Novartis Vaccines", "Novartis",
         "Click Therapeutics, Inc.", "Eli Lilly and Company",
         "Merck Sharp & Dohme LLC", "Hoffmann-La Roche",
         "Pfizer's Upjohn has merged with Mylan to form Viatris Inc.",
         "Metsera, a wholly owned subsidiary of Pfizer"]
for t in tests:
    print(f"  '{t}' → '{clean_sponsor_v3(t)}'")
```

```python
# PURPOSE: Handle weird edge cases with a manual map, then apply to all 5000 rows.
# WHY: Some sponsor names are sentences or contain subsidiary notes.
# These are too rare and weird for rules — just map them by hand.
# Like: most clothes go in the washing machine, but silk needs hand wash.

SPONSOR_MANUAL_MAP = {
    "Pfizer's Upjohn has merged with Mylan to form Viatris Inc.": "Viatris",
    "Metsera, a wholly owned subsidiary of Pfizer": "Pfizer",
    "Wyeth is now a wholly owned subsidiary of Pfizer": "Pfizer",
}

def clean_sponsor_final(name):
    """Final sponsor cleaner: manual map → suffix stripping."""
    if pd.isna(name):
        return name
    
    trimmed = name.strip()
    
    # Step 1: Check manual map first
    if trimmed in SPONSOR_MANUAL_MAP:
        return SPONSOR_MANUAL_MAP[trimmed]
    
    # Step 2: Apply suffix stripping
    return clean_sponsor_v3(trimmed)

# Apply to whole dataset
df['sponsor_cleaned'] = df['sponsor'].apply(clean_sponsor_final)

print(f"Before: {df['sponsor'].nunique()} unique sponsors")
print(f"After:  {df['sponsor_cleaned'].nunique()} unique sponsors")
print(f"Reduced by: {df['sponsor'].nunique() - df['sponsor_cleaned'].nunique()}")
```

```python
# PURPOSE: Standardize all corporate suffix variations to ONE canonical form.
# WHY: "Inc.", "Inc", "Incorporated" are the same thing written differently.
# We pick one spelling and force all to match.
# Like making everyone write dates as DD/MM/YYYY instead of mixing formats.

import re

def standardize_sponsor_v3(name):
    """Standardize sponsor suffix variations to canonical forms."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # 1. Standardize "Co., Ltd" variations → "Co Ltd"
    #    Handles: Co., Ltd / Co. Ltd / Co.,Ltd / Co.,Ltd. / Co., Ltd.
    cleaned = re.sub(r'\bCo\.\s*,?\s*Ltd\.?', 'Co Ltd', cleaned)
    
    # 2. Standardize Incorporated → Inc
    cleaned = re.sub(r',?\s*Incorporated\s*$', ' Inc', cleaned)
    
    # 3. Standardize "Inc." or ", Inc." or ", Inc" → " Inc"
    cleaned = re.sub(r',?\s*Inc\.?\s*$', ' Inc', cleaned)
    
    # 4. Standardize Limited / LTD → Ltd
    cleaned = re.sub(r',?\s*Limited\s*$', ' Ltd', cleaned)
    cleaned = re.sub(r',?\s*LTD\.?\s*$', ' Ltd', cleaned)
    cleaned = re.sub(r',?\s*Ltd\.?\s*$', ' Ltd', cleaned)
    
    # 5. Standardize Corporation / Corp. → Corp
    cleaned = re.sub(r',?\s*Corporation\s*$', ' Corp', cleaned)
    cleaned = re.sub(r',?\s*Corp\.?\s*$', ' Corp', cleaned)
    
    # 6. Standardize LLC. → LLC
    cleaned = re.sub(r',?\s*LLC\.?\s*$', ' LLC', cleaned)
    
    # 7. Standardize Pharmaceuticals / Pharmaceutical → Pharma
    #cleaned = re.sub(r',?\s*Pharmaceuticals?\s*$', ' Pharma', cleaned)
    
    # 8. Standardize Pty Ltd → Ltd
    cleaned = re.sub(r',?\s*Pty\s+Ltd\.?\s*$', ' Ltd', cleaned)
    
    # 9. Standardize gGmbH → GmbH
    cleaned = re.sub(r'\bgGmbH\s*$', 'GmbH', cleaned)
    
    # 10. Standardize S.p.A. variations
    cleaned = re.sub(r',?\s*S\.?p\.?A\.?\s*$', ' SpA', cleaned)
    
    # 11. Standardize B.V. → BV
    cleaned = re.sub(r',?\s*B\.?V\.?\s*$', ' BV', cleaned)
    
    # 12. Standardize N.V. → NV
    cleaned = re.sub(r',?\s*N\.?V\.?\s*$', ' NV', cleaned)
    
    # Clean up extra spaces
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    return cleaned

# Test
tests = [
    "Milestone Pharmaceuticals Inc.",
    "Novartis Pharmaceuticals",
    "Click Therapeutics, Inc.",
    "Celltex Therapeutics Corporation",
    "SK Chemicals Co., Ltd.",
    "Eli Lilly and Company",
    "Merck Sharp & Dohme LLC",
    "Staidson Co.,Ltd",
]
for t in tests:
    print(f"  '{t}' → '{standardize_sponsor_v3(t)}'")
```

```python
# PURPOSE: Extract the "core" of a sponsor name by stripping ALL suffixes.
# WHY: We only use this core for COMPARING names, not as the final output.
# The final output keeps the standardized suffix.
# Like comparing two people by face (core) not by their hat (suffix).

SUFFIXES_TO_STRIP_FOR_CORE = [
    r'\bCo\s*Ltd$',
    r'\bInc$',
    r'\bLtd$',
    r'\bLLC$',
    r'\bCorp$',
    r'\bGmbH$',
    r'\bSpA$',
    r'\bBV$',
    r'\bNV$',
    r'\bSA$',
    r'\bSL$',
    r'\bSAS$',
    r'\bAG$',
    r'\bAB$',
    r'\bA/S$',
    r'\bPLC$',
    r'\bSE$',
    r'\bPLLC$',
    r'\band Company$',
    r'\b& Company$',
]

def get_core_name(name):
    """Strip all suffixes to get the core name for comparison only."""
    if pd.isna(name):
        return name
    
    core = name.strip()
    
    for suffix in SUFFIXES_TO_STRIP_FOR_CORE:
        core = re.sub(suffix, '', core, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation and extra spaces
    core = core.rstrip('.,;: ')
    core = re.sub(r'\s+', ' ', core).strip()
    
    return core

# Test: compare core names
test_pairs = [
    ("Pfizer Inc", "Pfizer"),
    ("Novartis Pharmaceuticals", "Novartis"),
    ("Cairo University", "Cairo University Hospital"),
    ("Eli Lilly and Company", "Eli Lilly"),
    ("Merck Sharp & Dohme LLC", "Merck Sharp & Dohme"),
    ("University of California, San Francisco", "University of California San Francisco"),
]

for a, b in test_pairs:
    core_a = get_core_name(a)
    core_b = get_core_name(b)
    score = fuzz.token_sort_ratio(core_a, core_b)
    print(f"  '{a}' → core: '{core_a}'")
    print(f"  '{b}' → core: '{core_b}'")
    print(f"    score: {score}")
    print()
```

```python
# PURPOSE: Add division words like "Pharmaceuticals", "Therapeutics" etc.
# to the core stripper FOR COMPARISON ONLY.
# WHY: "Novartis Pharmaceuticals" and "Novartis" are the same company.
# We strip "Pharmaceuticals" only when comparing, NOT in the final output.
# Like ignoring someone's job title when checking if they're the same person.

SUFFIXES_TO_STRIP_FOR_CORE = [
    # Corporate legal suffixes
    r'\bCo\s*Ltd$',
    r'\bInc$',
    r'\bLtd$',
    r'\bLLC$',
    r'\bCorp$',
    r'\bGmbH$',
    r'\bSpA$',
    r'\bBV$',
    r'\bNV$',
    r'\bSA$',
    r'\bSL$',
    r'\bSAS$',
    r'\bAG$',
    r'\bAB$',
    r'\bA/S$',
    r'\bPLC$',
    r'\bSE$',
    r'\bPLLC$',
    r'\band Company$',
    r'\b& Company$',
    # Division words
    r'\bPharmaceuticals?$',
    r'\bPharma$',
    r'\bVaccines?$',
    r'\bBiosciences?$',
    r'\bBiotech$',
    r'\bTherapeutics?$',
    r'\bBiopharma$',
    r'\bBiopharmaceuticals?$',
]

def get_core_name(name):
    """Strip all suffixes to get core name for comparison only."""
    if pd.isna(name):
        return name
    
    core = name.strip()
    
    # Run twice to catch combos like "Therapeutics Corp"
    for _ in range(2):
        for suffix in SUFFIXES_TO_STRIP_FOR_CORE:
            core = re.sub(suffix, '', core, flags=re.IGNORECASE).strip()
    
    core = core.rstrip('.,;: ')
    core = re.sub(r'\s+', ' ', core).strip()
    
    return core

# Test again
test_pairs = [
    ("Novartis Pharmaceuticals", "Novartis"),
    ("Novartis Vaccines", "Novartis"),
    ("Novartis Vaccines", "Novartis Pharmaceuticals"),
    ("Cairo University", "Cairo University Hospital"),
    ("Celltex Therapeutics Corp", "Celltex"),
    ("Samyang Biopharmaceuticals Corp", "Samyang"),
]

for a, b in test_pairs:
    core_a = get_core_name(a)
    core_b = get_core_name(b)
    score = fuzz.token_sort_ratio(core_a, core_b)
    print(f"  '{a}' → core: '{core_a}'")
    print(f"  '{b}' → core: '{core_b}'")
    print(f"    score: {score}")
    print()
```

```python
# PURPOSE: Group sponsors by their core name, then pick the most frequent
# version as the canonical name. Map all variations to it.
# WHY: "Novartis Vaccines" and "Novartis Pharmaceuticals" share core "Novartis"
# → they're the same company → map both to the most common version.
# Like grouping all mail for the Smith family into one mailbox.

from collections import defaultdict

# Step 1: Get all unique cleaned sponsors and their core names
unique_sponsors = df['sponsor_cleaned'].unique()
sponsor_cores = {s: get_core_name(s).lower() for s in unique_sponsors}

# Step 2: Group by exact core match
core_groups = defaultdict(list)
for sponsor, core in sponsor_cores.items():
    core_groups[core].append(sponsor)

# Step 3: Pick most frequent as canonical, build mapping
sponsor_map = {}
groups_found = 0

for core, members in core_groups.items():
    if len(members) > 1:
        groups_found += 1
        # Count how often each version appears in data
        freqs = {m: (df['sponsor_cleaned'] == m).sum() for m in members}
        canonical = max(freqs, key=freqs.get)
        for m in members:
            if m != canonical:
                sponsor_map[m] = canonical

print(f"Groups with multiple versions: {groups_found}")
print(f"Sponsors to remap: {len(sponsor_map)}")
print(f"\nAll mappings:")
for old, new in sorted(sponsor_map.items()):
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Fuzzy match core names to find more groups, then pick shortest
# name as canonical with the most common suffix attached.
# WHY: Exact core matching only found 5 groups. Fuzzy matching catches
# near-duplicates like slight spelling differences.
# Like sorting mail: even if "Jon Smith" and "John Smith" are slightly 
# different, they probably live in the same house.

from collections import defaultdict
from thefuzz import fuzz

# Step 1: Get unique cores
unique_sponsors = df['sponsor_cleaned'].unique()
sponsor_cores = {s: get_core_name(s) for s in unique_sponsors}

# Get unique cores and their members
core_to_sponsors = defaultdict(list)
for sponsor, core in sponsor_cores.items():
    core_to_sponsors[core.lower()].append(sponsor)

# Step 2: Fuzzy match between cores (threshold 90)
unique_cores = list(core_to_sponsors.keys())
merged_cores = {}  # maps core → group leader core

for i, core_a in enumerate(unique_cores):
    if core_a in merged_cores:
        continue
    for j in range(i + 1, len(unique_cores)):
        core_b = unique_cores[j]
        if core_b in merged_cores:
            continue
        score = fuzz.token_sort_ratio(core_a, core_b)
        if score >= 90:
            merged_cores[core_b] = core_a

# Step 3: Build final groups
final_groups = defaultdict(list)
for core, sponsors in core_to_sponsors.items():
    leader = merged_cores.get(core, core)
    final_groups[leader].extend(sponsors)

# Step 4: Pick shortest name as canonical
sponsor_map = {}
groups_found = 0

for leader, members in final_groups.items():
    if len(members) > 1:
        groups_found += 1
        canonical = min(members, key=len)
        for m in members:
            if m != canonical:
                sponsor_map[m] = canonical

print(f"Groups with multiple versions: {groups_found}")
print(f"Sponsors to remap: {len(sponsor_map)}")
print(f"\nAll mappings:")
for old, new in sorted(sponsor_map.items()):
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Fuzzy match ONLY pharma/biotech companies, not universities/hospitals.
# WHY: Universities all look similar ("University of X") causing false merges.
# Pharma companies have unique names so fuzzy matching is safe.
# Like: you can sort candy by color, but don't sort identical marbles by color.

from collections import defaultdict
from thefuzz import fuzz

# Keywords that indicate an institution (skip fuzzy for these)
INSTITUTION_KEYWORDS = [
    'university', 'hospital', 'clinic', 'institute', 'college',
    'medical center', 'cancer center', 'health system', 'school',
    'foundation', 'ministry', 'department', 'va ', 'veterans',
    'national', 'centre', 'center', 'nhs',
]

def is_institution(name):
    """Check if sponsor is a university/hospital/institution."""
    lower = name.lower()
    return any(kw in lower for kw in INSTITUTION_KEYWORDS)

# Step 1: Get cores and split into pharma vs institution
unique_sponsors = df['sponsor_cleaned'].unique()
sponsor_cores = {s: get_core_name(s) for s in unique_sponsors}

core_to_sponsors = defaultdict(list)
for sponsor, core in sponsor_cores.items():
    core_to_sponsors[core.lower()].append(sponsor)

# Step 2: Separate pharma cores from institution cores
pharma_cores = {}
institution_cores = {}
for core, sponsors in core_to_sponsors.items():
    if any(is_institution(s) for s in sponsors):
        institution_cores[core] = sponsors
    else:
        pharma_cores[core] = sponsors

print(f"Pharma/biotech cores: {len(pharma_cores)}")
print(f"Institution cores: {len(institution_cores)}")

# Step 3: Fuzzy match ONLY pharma cores (threshold 90)
pharma_core_list = list(pharma_cores.keys())
merged_cores = {}

for i, core_a in enumerate(pharma_core_list):
    if core_a in merged_cores:
        continue
    for j in range(i + 1, len(pharma_core_list)):
        core_b = pharma_core_list[j]
        if core_b in merged_cores:
            continue
        score = fuzz.token_sort_ratio(core_a, core_b)
        if score >= 90:
            merged_cores[core_b] = core_a

# Step 4: Build final groups (pharma fuzzy + institution exact)
final_groups = defaultdict(list)

# Pharma: use fuzzy groups
for core, sponsors in pharma_cores.items():
    leader = merged_cores.get(core, core)
    final_groups[leader].extend(sponsors)

# Institution: exact core only
for core, sponsors in institution_cores.items():
    final_groups[core].extend(sponsors)

# Step 5: Pick shortest name as canonical
sponsor_map = {}
groups_found = 0

for leader, members in final_groups.items():
    if len(members) > 1:
        groups_found += 1
        canonical = min(members, key=len)
        for m in members:
            if m != canonical:
                sponsor_map[m] = canonical

print(f"\nGroups with multiple versions: {groups_found}")
print(f"Sponsors to remap: {len(sponsor_map)}")
print(f"\nAll mappings:")
for old, new in sorted(sponsor_map.items()):
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Apply the sponsor mapping to the dataset and see final results.
# WHY: We've verified all 8 mappings look correct. Time to apply.

# Apply: first standardize suffixes, then remap groups
def clean_sponsor_complete(name):
    """Full pipeline: manual map → standardize suffix → group remap."""
    if pd.isna(name):
        return name
    
    trimmed = name.strip()
    
    # Step 1: Manual map (weird sentences)
    if trimmed in SPONSOR_MANUAL_MAP:
        trimmed = SPONSOR_MANUAL_MAP[trimmed]
    
    # Step 2: Standardize suffixes
    cleaned = standardize_sponsor_v3(trimmed)
    
    # Step 3: Group remap
    if cleaned in sponsor_map:
        cleaned = sponsor_map[cleaned]
    
    return cleaned

df['sponsor_cleaned'] = df['sponsor'].apply(clean_sponsor_complete)

before = df['sponsor'].nunique()
after = df['sponsor_cleaned'].nunique()
print(f"Before: {before} unique sponsors")
print(f"After:  {after} unique sponsors")
print(f"Reduced by: {before - after}")
```

```python
# PURPOSE: Build a function to clean individual drug names.
# WHAT IT DOES:
#   1) Strip dosage info (numbers + mg/mcg/ml/%)
#   2) Extract aliases from parentheses (Chinese or regular)
#   3) Normalize casing to Title Case
#   4) Clean whitespace
# WHY: Get names to a consistent base form before API lookups.
# Like removing packaging before comparing products.

import re

def clean_drug_name(name):
    """Clean a single drug name (not the whole cell, just one drug)."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Remove dosage patterns like "10 mg", "500mg", "1.5 mcg/kg", "0.9%"
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    
    # Step 2: Remove "daily", "weekly", "twice daily" etc.
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 3: Remove "in X ML Injection" type patterns
    cleaned = re.sub(r'\bin\s+\d+\s*ML\s+\w+', '', cleaned, flags=re.IGNORECASE)
    
    # Step 4: Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    # Step 5: Title case (but preserve ALL-CAPS abbreviations like "CAR" "BCMA")
    if cleaned and not cleaned.isupper():
        cleaned = cleaned.title()
    
    # Step 6: Remove trailing/leading punctuation
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "Henagliflozin 10 mg daily",
    "Aspirin 81mg",
    "CR845 0.5 mcg/kg",
    "Bridion 200 MG in 2 ML Injection",
    "Pembrolizumab",
    "pembrolizumab",
    "KEYTRUDA",
    "Placebo",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name(t)}'")
```

```python
# PURPOSE: Fix two issues:
#   1) "Bridion In Injection" — remove orphaned words after dosage stripping
#   2) "KEYTRUDA" — title case ALL-CAPS names too (they're brand names)
# WHY: Dosage removal leaves behind words like "in", "Injection", "Capsule".
# And brand names in ALL CAPS should become Title Case for consistency.

def clean_drug_name_v2(name):
    """Improved drug name cleaner."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Remove dosage patterns
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    
    # Step 2: Remove pharmaceutical form words (leftover after dosage removal)
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    
    # Step 3: Remove time/frequency words
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 4: Remove orphaned small words ("in", "of", "for") at boundaries
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 5: Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    # Step 6: Title Case everything
    if cleaned:
        cleaned = cleaned.title()
    
    # Step 7: Remove trailing/leading punctuation
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "Bridion 200 MG in 2 ML Injection",
    "KEYTRUDA",
    "pembrolizumab",
    "Henagliflozin 10 mg daily",
    "Aspirin 81mg",
    "Placebo",
    "Sodium Oligomannate Capsules",
    "CR845 0.5 mcg/kg",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v2(t)}'")
```

```python
# PURPOSE: Fix Title Case breaking research codes like "CR845" → "Cr845".
# WHY: Research codes are alphanumeric (letters + numbers) like "CR845", 
# "ABT-089", "MK-3475". Title Case ruins them.
# RULE: If a word has letters AND numbers mixed, keep it UPPERCASE.
# Like: don't correct a license plate's spelling.

def smart_title_case(text):
    """Title case but preserve research codes (letters + numbers mixed)."""
    words = text.split()
    result = []
    for word in words:
        # If word mixes letters and digits → keep uppercase
        has_letter = any(c.isalpha() for c in word)
        has_digit = any(c.isdigit() for c in word)
        if has_letter and has_digit:
            result.append(word.upper())
        else:
            result.append(word.title())
    return ' '.join(result)

def clean_drug_name_v3(name):
    """Drug name cleaner with smart casing."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Remove dosage patterns
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    
    # Step 2: Remove pharmaceutical form words
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    
    # Step 3: Remove time/frequency words
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 4: Remove orphaned small words
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 5: Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    # Step 6: Smart Title Case
    if cleaned:
        cleaned = smart_title_case(cleaned)
    
    # Step 7: Remove trailing/leading punctuation
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "CR845 0.5 mcg/kg",
    "ABT-089",
    "MK-3475",
    "TAS-117",
    "KEYTRUDA",
    "pembrolizumab",
    "Bridion 200 MG in 2 ML Injection",
    "Sodium Oligomannate Capsules",
    "5-Amino Levulinic Acid",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v3(t)}'")
```

```python
# PURPOSE: Fix "5-Amino" being treated as a research code.
# WHY: "5-Amino" has digit + letters but it's NOT a code — it's a chemical prefix.
# RULE: Only keep uppercase if digits and letters are MIXED together
# like "CR845", "ABT089". Not if separated by hyphen like "5-Amino".
# Better check: the word without hyphens still mixes letters+digits in same chunk.

def smart_title_case_v2(text):
    """Title case but preserve true research codes."""
    words = text.split()
    result = []
    for word in words:
        # Split by hyphen and check each part
        parts = word.split('-')
        is_code = False
        for part in parts:
            has_letter = any(c.isalpha() for c in part)
            has_digit = any(c.isdigit() for c in part)
            if has_letter and has_digit:
                is_code = True
                break
        
        if is_code:
            result.append(word.upper())
        else:
            result.append(word.title())
    return ' '.join(result)

def clean_drug_name_v4(name):
    """Drug name cleaner with improved casing."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    if cleaned:
        cleaned = smart_title_case_v2(cleaned)
    
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "5-Amino Levulinic Acid",
    "CR845 0.5 mcg/kg",
    "ABT-089",
    "MK-3475",
    "KEYTRUDA",
    "pembrolizumab",
    "Bridion 200 MG in 2 ML Injection",
    "IL-12",
    "SHR-1210",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v4(t)}'")
```

```python
# PURPOSE: Better rule for research codes.
# INSIGHT: Research codes follow a pattern: 2-4 UPPERCASE LETTERS + hyphen + DIGITS
# like ABT-089, MK-3475, SHR-1210, IL-12, CR845.
# If the ORIGINAL word was ALL UPPERCASE and short, it's likely a code.
# RULE: If the original word is ALL CAPS (or all caps + digits), keep uppercase.

def smart_title_case_v3(text, original):
    """Title case but preserve research codes based on original casing."""
    words = text.split()
    orig_words = original.split()
    
    result = []
    for i, word in enumerate(words):
        # Find matching original word if possible
        orig = orig_words[i] if i < len(orig_words) else word
        
        # Rule 1: If original was ALL UPPERCASE → keep uppercase
        # Catches: ABT-089, MK-3475, KEYTRUDA, IL-12, SHR-1210, CR845
        if orig.replace('-', '').replace('.', '').isupper() and any(c.isalpha() for c in orig):
            result.append(word.upper())
        else:
            result.append(word.title())
    
    return ' '.join(result)

def clean_drug_name_v5(name):
    """Drug name cleaner with proper code detection."""
    if pd.isna(name) or not name.strip():
        return name
    
    original = name.strip()
    cleaned = original
    
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    cleaned = cleaned.strip('.,;:- ')
    
    if not cleaned:
        return original
    
    # Smart casing: compare each word to original
    cleaned = smart_title_case_v3(cleaned, original)
    
    return cleaned

# Test
tests = [
    "ABT-089", "MK-3475", "SHR-1210", "IL-12", "CR845",
    "KEYTRUDA", "pembrolizumab", "5-Amino Levulinic Acid",
    "Bridion 200 MG in 2 ML Injection",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v5(t)}'")
```

```python
# PURPOSE: Apply clean_drug_name_v5 to every row.
# WHY: Remember, many cells have multiple drugs separated by "+".
# So we: split by "+" → clean each drug → rejoin.
# Same approach as country (split, clean, rejoin).

def clean_drug_cell(cell):
    """Clean a cell that may contain multiple drugs separated by +."""
    if pd.isna(cell):
        return cell
    
    # Split by "+"
    drugs = cell.split('+')
    
    # Clean each drug individually
    cleaned = [clean_drug_name_v5(d.strip()) for d in drugs]
    
    # Remove empty strings and rejoin
    cleaned = [d for d in cleaned if d.strip()]
    
    return ' + '.join(cleaned)

# Apply to all rows
df['intervention_cleaned'] = df['intervention_name'].apply(clean_drug_cell)

print(f"Before: {df['intervention_name'].nunique()} unique")
print(f"After:  {df['intervention_cleaned'].nunique()} unique")
print(f"Reduced by: {df['intervention_name'].nunique() - df['intervention_cleaned'].nunique()}")

# Sample changes
changed = df[df['intervention_name'] != df['intervention_cleaned']][['intervention_name', 'intervention_cleaned']]
print(f"\nRows changed: {len(changed)}")
print(f"\nSample changes:")
for _, row in changed.sample(10, random_state=42).iterrows():
    print(f"  '{row['intervention_name']}'")
    print(f"  → '{row['intervention_cleaned']}'")
    print()
```

```python
# PURPOSE: Remove the orphaned small word stripping — it causes more harm than good.
# WHY: "Placebo of DA-2803" → "Placebo DA-2803" is wrong.
# "of", "in", "for" are meaningful in drug names.
# Also add "QD" (once daily), "BID" (twice daily) to frequency stripping.

def clean_drug_name_v6(name):
    """Drug name cleaner — conservative, no small word stripping."""
    if pd.isna(name) or not name.strip():
        return name
    
    original = name.strip()
    cleaned = original
    
    # Step 1: Remove dosage patterns
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    
    # Step 2: Remove pharmaceutical forms
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    
    # Step 3: Remove frequency words
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week|QD|BID|TID|QID)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 4: Normalize whitespace (NO small word stripping)
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    cleaned = cleaned.strip('.,;:- ')
    
    if not cleaned:
        return original
    
    # Step 5: Smart casing
    cleaned = smart_title_case_v3(cleaned, original)
    
    return cleaned

def clean_drug_cell_v2(cell):
    """Clean multi-drug cell."""
    if pd.isna(cell):
        return cell
    
    drugs = cell.split('+')
    cleaned = [clean_drug_name_v6(d.strip()) for d in drugs]
    cleaned = [d for d in cleaned if d.strip()]
    
    return ' + '.join(cleaned)

# Test the problem cases
tests = [
    "Placebo of DA-2803",
    "DA-2803-R",
    "CJ-12420 100 mg QD",
    "Esomeprazole 40 mg QD",
    "KEYTRUDA",
    "pembrolizumab",
    "ABT-089",
    "5-Amino Levulinic Acid",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v6(t)}'")
```

```python
# PURPOSE: Filter the 3629 names down to ones that LOOK like real drugs.
# WHY: "0.4Mu Insulin Followed By..." is not a drug name.
# RULES for a likely drug name:
#   - Starts with a letter (not a number)
#   - Between 3 and 50 characters
#   - Not a study description (no "followed by", "arm", "dose of")
#   - Not just a common non-drug word
# Like sorting a pile of mail: toss the junk, keep the real letters.

NON_DRUG_PATTERNS = [
    r'^[\d]',                    # starts with number
    r'followed by',              # study design language
    r'\barm\b',                  # study arm
    r'\bdose of\b',              # dosage description
    r'\bstandard\b.*\bcare\b',   # "standard of care"
    r'\bplacebo\b',              # placebo
    r'\bcontrol\b',              # control group
    r'\bsham\b',                 # sham procedure
    r'\bno intervention\b',      # no intervention
    r'\busual care\b',           # usual care
    r'\bwithout\b',              # exclusion language
]

def is_likely_drug(name):
    """Check if a name looks like a real drug name."""
    if not name or len(name) < 3 or len(name) > 50:
        return False
    if not name[0].isalpha():
        return False
    lower = name.lower()
    for pattern in NON_DRUG_PATTERNS:
        if re.search(pattern, lower):
            return False
    return True

likely_drugs = sorted([d for d in all_drugs if is_likely_drug(d)])
print(f"Before filter: {len(all_drugs)}")
print(f"After filter:  {len(likely_drugs)}")
print(f"\nSample 20:")
for d in likely_drugs[:20]:
    print(f"  {d}")
```

```python
# PURPOSE: Test the RxNorm API with a few known drug names.
# WHY: Before running 3189 calls, make sure the API works and returns
# what we expect. Like testing a key in one lock before copying it.
# 
# RxNorm API is FREE, no key needed. It maps drug names to a 
# canonical "RxCUI" (unique ID) and returns the generic name.
# Endpoint: https://rxnav.nlm.nih.gov/REST/rxcui.json?name=DRUGNAME

import requests

def rxnorm_lookup(drug_name):
    """Look up a drug name in RxNorm. Returns generic name or None."""
    url = "https://rxnav.nlm.nih.gov/REST/rxcui.json"
    params = {"name": drug_name}
    
    try:
        resp = requests.get(url, params=params, timeout=10)
        data = resp.json()
        
        # Get RxCUI (unique drug ID)
        ids = data.get("idGroup", {}).get("rxnormId", [])
        if not ids:
            return None
        
        rxcui = ids[0]
        
        # Get properties (including generic name)
        prop_url = f"https://rxnav.nlm.nih.gov/REST/rxcui/{rxcui}/properties.json"
        prop_resp = requests.get(prop_url, timeout=10)
        prop_data = prop_resp.json()
        
        props = prop_data.get("properties", {})
        return props.get("name", None)
    
    except Exception as e:
        return f"ERROR: {e}"

# Test with known drugs
tests = ["Pembrolizumab", "KEYTRUDA", "MK-3475", "Aspirin", 
         "Metformin", "Nivolumab", "Ibuprofen"]
for t in tests:
    result = rxnorm_lookup(t)
    print(f"  '{t}' → '{result}'")
```

```python
# PURPOSE: Try RxNorm's "approximate match" and "related" endpoints.
# WHY: Basic lookup just confirms a name exists. We need:
#   - Brand → Generic mapping (KEYTRUDA → Pembrolizumab)
#   - Research code → Generic mapping (MK-3475 → Pembrolizumab)
# RxNorm has a "getAllRelatedInfo" endpoint that shows relationships.

def rxnorm_to_generic(drug_name):
    """Try to find the generic/ingredient name for any drug name."""
    
    # Step 1: Try approximate match to get RxCUI
    url = "https://rxnav.nlm.nih.gov/REST/approximateTerm.json"
    params = {"term": drug_name, "maxEntries": 1}
    
    try:
        resp = requests.get(url, params=params, timeout=10)
        data = resp.json()
        
        candidates = data.get("approximateGroup", {}).get("candidate", [])
        if not candidates:
            return None
        
        rxcui = candidates[0].get("rxcui")
        if not rxcui:
            return None
        
        # Step 2: Get related ingredients (generic names)
        rel_url = f"https://rxnav.nlm.nih.gov/REST/rxcui/{rxcui}/related.json"
        rel_params = {"tty": "IN"}  # IN = Ingredient (generic name)
        rel_resp = requests.get(rel_url, params=rel_params, timeout=10)
        rel_data = rel_resp.json()
        
        # Extract ingredient names
        groups = rel_data.get("relatedGroup", {}).get("conceptGroup", [])
        for group in groups:
            props = group.get("conceptProperties", [])
            if props:
                return props[0].get("name")
        
        # If no ingredient found, return the name from approximate match
        return candidates[0].get("name")
    
    except Exception as e:
        return f"ERROR: {e}"

# Test
tests = ["KEYTRUDA", "Pembrolizumab", "MK-3475", "Aspirin",
         "Opdivo", "Nivolumab", "Herceptin", "Metformin"]
for t in tests:
    result = rxnorm_to_generic(t)
    print(f"  '{t}' → '{result}'")
```

```python
# PURPOSE: Reduce API calls by filtering smarter.
# WHY: 65 minutes is too long. Most names won't be in RxNorm anyway.
# STRATEGY: 
#   - Only DRUG/BIOLOGICAL types
#   - Skip names longer than 30 chars (likely descriptions, not drug names)
#   - Skip names with more than 3 words (likely descriptions)
#   - Count how many remain
# Like only calling the doctor for real symptoms, not every itch.

def is_rxnorm_candidate(name):
    """Check if name is worth looking up in RxNorm."""
    if not name or len(name) < 2:
        return False
    # Too long = probably a description
    if len(name) > 30:
        return False
    # Too many words = probably a description
    if len(name.split()) > 3:
        return False
    # Starts with number = probably dosage fragment
    if not name[0].isalpha():
        return False
    # Common non-drug terms
    skip_terms = ['placebo', 'sham', 'control', 'usual care', 'no intervention',
                  'standard care', 'exercise', 'diet', 'counseling', 'education',
                  'surgery', 'physiotherapy', 'questionnaire', 'interview']
    if name.lower() in skip_terms:
        return False
    return True

candidates = sorted([d for d in likely_drugs if is_rxnorm_candidate(d)])
print(f"From {len(likely_drugs)} likely drugs → {len(candidates)} RxNorm candidates")
print(f"Estimated time: {len(candidates) * 1.22 / 60:.1f} minutes")
print(f"\nSample 20:")
for d in candidates[:20]:
    print(f"  {d}")
```

```python
# PURPOSE: Smarter filtering to remove obvious non-drugs.
# WHY: "A Painbuster® Catheter" is a device, "ACT Guided Heparinization" 
# is a procedure. We want only actual drug/compound names.
# RULES:
#   - Must be DRUG or BIOLOGICAL intervention type
#   - No special characters (®, ™, ©)
#   - No procedure/device words (catheter, block, surgery, device, etc.)
#   - Skip pure abbreviations less than 3 chars
#   - Max 4 words (real drug names are short)

NON_DRUG_WORDS = [
    'catheter', 'device', 'block', 'surgery', 'procedure', 'stent',
    'implant', 'pump', 'monitor', 'mask', 'bandage', 'dressing',
    'exercise', 'training', 'therapy', 'counseling', 'education',
    'program', 'intervention', 'questionnaire', 'interview', 'app',
    'guided', 'assisted', 'based', 'conventional', 'standard',
    'system', 'technique', 'method', 'protocol', 'regimen',
    'diet', 'meal', 'food', 'supplement', 'vitamin',
    'sham', 'control', 'placebo', 'usual care', 'no intervention',
    'needle', 'suture', 'graft', 'prosthesis', 'scaffold',
]

def is_drug_candidate(name):
    """Strict filter: is this likely an actual drug/compound name?"""
    if not name or len(name) < 3:
        return False
    if not name[0].isalpha():
        return False
    if len(name) > 40:
        return False
    if len(name.split()) > 4:
        return False
    # Special characters = likely device/brand product
    if any(c in name for c in '®™©'):
        return False
    # Check for non-drug words
    lower = name.lower()
    for word in NON_DRUG_WORDS:
        if word in lower:
            return False
    return True

# Collect unique drug names from DRUG/BIOLOGICAL rows only
drug_mask = df['intervention_type'].str.contains('DRUG|BIOLOGICAL', na=False)
all_individual_drugs = set()
for cell in df[drug_mask]['intervention_cleaned'].dropna():
    for drug in cell.split('+'):
        name = drug.strip()
        if name and is_drug_candidate(name):
            all_individual_drugs.add(name)

candidates = sorted(all_individual_drugs)
print(f"Final candidates for RxNorm: {len(candidates)}")
print(f"Estimated time: {len(candidates) * 1.22 / 60:.1f} minutes")
print(f"\nFirst 30:")
for d in candidates[:30]:
    print(f"  {d}")
```

```python
# PURPOSE: Remove bad mappings from the drug dictionary.
# WHY: RxNorm's approximate match sometimes returns nonsense.
# RULES for a GOOD mapping:
#   1) Input and output share at least some characters (not totally different)
#   2) Don't map combos to single drugs
#   3) Keep brand → generic (these are the gold)
# Like proofreading a translation dictionary — remove wrong translations.

from thefuzz import fuzz

clean_dict = {}
removed = []

for raw, generic in drug_dictionary.items():
    # Skip if raw contains combo separators (shouldn't have been looked up)
    if any(sep in raw for sep in [' And ', ' Plus ', ', ', '/']):
        removed.append((raw, generic, "combo string"))
        continue
    
    # Skip if mapping looks unrelated (fuzzy score too low AND no shared words)
    raw_words = set(raw.lower().split())
    gen_words = set(generic.lower().split())
    shared = raw_words & gen_words
    
    score = fuzz.ratio(raw.lower(), generic.lower())
    
    # Keep if: high similarity OR it's a known pattern (brand → generic)
    if score >= 30 or len(shared) > 0:
        clean_dict[raw] = generic
    else:
        removed.append((raw, generic, f"score={score}"))

print(f"Original dictionary: {len(drug_dictionary)}")
print(f"Cleaned dictionary: {len(clean_dict)}")
print(f"Removed: {len(removed)}")

print(f"\nRemoved mappings:")
for raw, gen, reason in removed:
    print(f"  '{raw}' → '{gen}' ({reason})")

print(f"\nSample good mappings (brand → generic):")
count = 0
for raw, generic in sorted(clean_dict.items()):
    if raw.lower() != generic.lower():
        print(f"  '{raw}' → '{generic}'")
        count += 1
        if count >= 15:
            break
```

```python
# PURPOSE: Build a function to clean individual condition names.
# STRATEGY:
#   1) Title Case normalization
#   2) Remove trailing plurals ("Diseases" → "Disease")
#   3) Abbreviation dictionary (NSCLC → Non-Small Cell Lung Cancer)
#   4) Normalize parenthetical abbreviations 
#      "Chronic Obstructive Pulmonary Disease (COPD)" → keep as canonical
#      "COPD" alone → expand to full name
#      "COPD (Chronic Obstructive Pulmonary Disease)" → flip to canonical form
# Like: everyone calls it "COPD" or the full name — we pick ONE form.

# Abbreviation → Full canonical name
CONDITION_ABBREVIATIONS = {
    'nsclc': 'Non-Small Cell Lung Cancer',
    'sclc': 'Small Cell Lung Cancer',
    'aml': 'Acute Myeloid Leukemia',
    'all': 'Acute Lymphoblastic Leukemia',
    'cll': 'Chronic Lymphocytic Leukemia',
    'nhl': 'Non-Hodgkin Lymphoma',
    'dlbcl': 'Diffuse Large B-Cell Lymphoma',
    'adhd': 'Attention Deficit Hyperactivity Disorder',
    'copd': 'Chronic Obstructive Pulmonary Disease',
    'ckd': 'Chronic Kidney Disease',
    'hcc': 'Hepatocellular Carcinoma',
    'rcc': 'Renal Cell Carcinoma',
    'gist': 'Gastrointestinal Stromal Tumor',
    'mds': 'Myelodysplastic Syndrome',
    'tnbc': 'Triple-Negative Breast Cancer',
    'gerd': 'Gastroesophageal Reflux Disease',
    'ibd': 'Inflammatory Bowel Disease',
    'ibs': 'Irritable Bowel Syndrome',
    'ptsd': 'Post-Traumatic Stress Disorder',
    'ocd': 'Obsessive-Compulsive Disorder',
    'ms': 'Multiple Sclerosis',
    'ra': 'Rheumatoid Arthritis',
    'sle': 'Systemic Lupus Erythematosus',
    'osa': 'Obstructive Sleep Apnea',
    'af': 'Atrial Fibrillation',
    'mi': 'Myocardial Infarction',
    'ards': 'Acute Respiratory Distress Syndrome',
    'nafld': 'Non-Alcoholic Fatty Liver Disease',
    'nash': 'Non-Alcoholic Steatohepatitis',
    'pcos': 'Polycystic Ovary Syndrome',
    'bph': 'Benign Prostatic Hyperplasia',
    'uti': 'Urinary Tract Infection',
    'dvt': 'Deep Vein Thrombosis',
    'pe': 'Pulmonary Embolism',
    'aki': 'Acute Kidney Injury',
    'tbi': 'Traumatic Brain Injury',
    'als': 'Amyotrophic Lateral Sclerosis',
    'hiv': 'HIV Infection',
    'hcv': 'Hepatitis C',
    'hbv': 'Hepatitis B',
    't2dm': 'Type 2 Diabetes Mellitus',
    't1dm': 'Type 1 Diabetes Mellitus',
    'ad': 'Alzheimer Disease',
    'pd': 'Parkinson Disease',
    'cf': 'Cystic Fibrosis',
    'uc': 'Ulcerative Colitis',
    'cd': 'Crohn Disease',
    'adpkd': 'Autosomal Dominant Polycystic Kidney Disease',
}

def clean_condition(name):
    """Clean a single condition name."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Check if it's a pure abbreviation (e.g., "NSCLC", "COPD")
    lower = cleaned.lower().strip()
    if lower in CONDITION_ABBREVIATIONS:
        return CONDITION_ABBREVIATIONS[lower]
    
    # Step 2: Remove parenthetical abbreviation at the end
    # "Chronic Obstructive Pulmonary Disease (COPD)" → "Chronic Obstructive Pulmonary Disease"
    cleaned = re.sub(r'\s*\([A-Z]{2,10}\)\s*$', '', cleaned).strip()
    
    # Step 3: Handle "ABBR - Full Name" or "ABBR (Full Name)" pattern
    # "DLBCL - Diffuse Large B Cell Lymphoma" → "Diffuse Large B-Cell Lymphoma"
    match = re.match(r'^([A-Z]{2,10})\s*[-–]\s*(.+)$', cleaned)
    if match:
        abbr = match.group(1).lower()
        if abbr in CONDITION_ABBREVIATIONS:
            return CONDITION_ABBREVIATIONS[abbr]
    
    # Step 4: Handle "ABBR (Full Name)" pattern
    match = re.match(r'^([A-Z]{2,10})\s*\((.+)\)\s*$', cleaned)
    if match:
        abbr = match.group(1).lower()
        if abbr in CONDITION_ABBREVIATIONS:
            return CONDITION_ABBREVIATIONS[abbr]
    
    # Step 5: Remove trailing plural 's' for common patterns
    # "Diseases" → "Disease", "Disorders" → "Disorder", "Syndromes" → "Syndrome"
    cleaned = re.sub(r'(Disease|Disorder|Syndrome|Infection|Neoplasm|Fracture|Tumor)s\b', 
                     r'\1', cleaned)
    
    # Step 6: Title Case (preserve known abbreviations)
    words = cleaned.split()
    result = []
    for word in words:
        # Keep ALL CAPS short words (likely abbreviations like HIV, ACL)
        if word.isupper() and len(word) <= 6 and word.isalpha():
            result.append(word)
        else:
            result.append(word.title() if word.islower() or word.isupper() else word)
    cleaned = ' '.join(result)
    
    # Step 7: Normalize common spelling variations
    cleaned = cleaned.replace('Leukaemia', 'Leukemia')
    cleaned = cleaned.replace('leukaemia', 'leukemia')
    cleaned = cleaned.replace('Tumour', 'Tumor')
    cleaned = cleaned.replace('tumour', 'tumor')
    cleaned = cleaned.replace('Haemophilia', 'Hemophilia')
    cleaned = cleaned.replace('Anaemia', 'Anemia')
    cleaned = cleaned.replace('Oesophageal', 'Esophageal')
    
    return cleaned

# Test
tests = [
    "NSCLC", "Non-Small Cell Lung Cancer (NSCLC)", "NSCLC Stage IV",
    "COPD", "Chronic Obstructive Pulmonary Disease (COPD)", "Copd",
    "COPD (Chronic Obstructive Pulmonary Disease)",
    "DLBCL - Diffuse Large B Cell Lymphoma", "DLBCL",
    "AML", "Acute Myeloid Leukemia (AML)", "AML (Acute Myeloid Leukemia)",
    "HIV Infections", "HIV INFECTIONS",
    "Myelodysplastic Syndromes", "Breast Cancer", "breast cancer",
    "Leukaemia (Acute Myeloid)",
]
for t in tests:
    print(f"  '{t}' → '{clean_condition(t)}'")
```

```python
# PURPOSE: Fix ordering — normalize casing first, THEN strip plurals.
# WHY: "HIV INFECTIONS" needs to become "HIV Infections" before
# the plural rule can match "Infections" → "Infection".
# Like: wash the mud off before checking for stains.

def clean_condition_v2(name):
    """Condition cleaner with fixed ordering."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Check if pure abbreviation
    lower = cleaned.lower().strip()
    if lower in CONDITION_ABBREVIATIONS:
        return CONDITION_ABBREVIATIONS[lower]
    
    # Step 2: Handle "ABBR - Full Name" pattern
    match = re.match(r'^([A-Z]{2,10})\s*[-–]\s*(.+)$', cleaned)
    if match:
        abbr = match.group(1).lower()
        if abbr in CONDITION_ABBREVIATIONS:
            return CONDITION_ABBREVIATIONS[abbr]
    
    # Step 3: Handle "ABBR (Full Name)" pattern
    match = re.match(r'^([A-Z]{2,10})\s*\((.+)\)\s*$', cleaned)
    if match:
        abbr = match.group(1).lower()
        if abbr in CONDITION_ABBREVIATIONS:
            return CONDITION_ABBREVIATIONS[abbr]
    
    # Step 4: Remove parenthetical abbreviation at the end
    cleaned = re.sub(r'\s*\([A-Z]{2,10}\)\s*$', '', cleaned).strip()
    
    # Step 5: Title Case FIRST (preserve short ALL-CAPS abbreviations)
    words = cleaned.split()
    result = []
    for word in words:
        if word.isupper() and len(word) <= 6 and word.isalpha():
            result.append(word)
        elif word.isupper():
            result.append(word.title())
        else:
            result.append(word[0].upper() + word[1:] if word else word)
    cleaned = ' '.join(result)
    
    # Step 6: THEN strip plurals
    cleaned = re.sub(r'(Disease|Disorder|Syndrome|Infection|Neoplasm|Fracture|Tumor|Ulcer)s\b',
                     r'\1', cleaned)
    
    # Step 7: Spelling normalization (British → American)
    cleaned = cleaned.replace('Leukaemia', 'Leukemia')
    cleaned = cleaned.replace('Tumour', 'Tumor')
    cleaned = cleaned.replace('Haemophilia', 'Hemophilia')
    cleaned = cleaned.replace('Anaemia', 'Anemia')
    cleaned = cleaned.replace('Oesophageal', 'Esophageal')
    cleaned = cleaned.replace('Behaviour', 'Behavior')
    cleaned = cleaned.replace('Colour', 'Color')
    
    return cleaned

# Test
tests = [
    "HIV INFECTIONS", "HIV Infections", "HIV Infection",
    "NSCLC Stage IV", "NSCLC",
    "COPD", "Copd", "Chronic Obstructive Pulmonary Disease (COPD)",
    "breast cancer", "BREAST CANCER", "Breast Cancer",
    "Myelodysplastic Syndromes", "Leukaemia (Acute Myeloid)",
    "Diabetes Mellitus, Type 2", "Type 2 Diabetes",
]
for t in tests:
    print(f"  '{t}' → '{clean_condition_v2(t)}'")
```

```python
# PURPOSE: Fix "BREAST CANCER" not getting title-cased.
# WHY: "BREAST" is 6 chars — my rule kept words ≤6 chars as uppercase.
# Better rule: only keep ALL-CAPS if it LOOKS like an abbreviation
# (no vowels pattern, or known abbreviation list).
# Real abbreviations: HIV, COPD, NSCLC, ACL, AML
# Not abbreviations: BREAST, CANCER, HEALTHY

KNOWN_ABBREVIATIONS = {
    'HIV', 'AIDS', 'COPD', 'NSCLC', 'SCLC', 'AML', 'ALL', 'CLL',
    'NHL', 'DLBCL', 'ADHD', 'CKD', 'HCC', 'RCC', 'GIST', 'MDS',
    'TNBC', 'GERD', 'IBD', 'IBS', 'PTSD', 'OCD', 'MS', 'RA',
    'SLE', 'OSA', 'AF', 'MI', 'ARDS', 'NAFLD', 'NASH', 'PCOS',
    'BPH', 'UTI', 'DVT', 'PE', 'AKI', 'TBI', 'ALS', 'HCV', 'HBV',
    'ACL', 'ICU', 'CAR', 'ALK', 'EGFR', 'BRCA', 'PD', 'AD',
    'ADPKD', 'EBV', 'HPV', 'CMV', 'RSV', 'TB', 'MRSA', 'GI',
    'CNS', 'CVD', 'PAD', 'VTE', 'HER2', 'BCG', 'IV', 'II', 'III',
}

def smart_title_case_condition(text):
    """Title case but preserve known medical abbreviations."""
    words = text.split()
    result = []
    for word in words:
        # Keep if it's a known abbreviation
        if word.upper() in KNOWN_ABBREVIATIONS:
            result.append(word.upper())
        # If ALL CAPS but not a known abbreviation → title case
        elif word.isupper() and len(word) > 1:
            result.append(word.title())
        # Already mixed case → keep first letter behavior
        else:
            result.append(word[0].upper() + word[1:] if word else word)
    return ' '.join(result)

def clean_condition_v3(name):
    """Condition cleaner with proper abbreviation handling."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Check pure abbreviation
    lower = cleaned.lower().strip()
    if lower in CONDITION_ABBREVIATIONS:
        return CONDITION_ABBREVIATIONS[lower]
    
    # Step 2: "ABBR - Full Name" pattern
    match = re.match(r'^([A-Z]{2,10})\s*[-–]\s*(.+)$', cleaned)
    if match:
        abbr = match.group(1).lower()
        if abbr in CONDITION_ABBREVIATIONS:
            return CONDITION_ABBREVIATIONS[abbr]
    
    # Step 3: "ABBR (Full Name)" pattern
    match = re.match(r'^([A-Z]{2,10})\s*\((.+)\)\s*$', cleaned)
    if match:
        abbr = match.group(1).lower()
        if abbr in CONDITION_ABBREVIATIONS:
            return CONDITION_ABBREVIATIONS[abbr]
    
    # Step 4: Remove trailing parenthetical abbreviation
    cleaned = re.sub(r'\s*\([A-Z]{2,10}\)\s*$', '', cleaned).strip()
    
    # Step 5: Smart Title Case
    cleaned = smart_title_case_condition(cleaned)
    
    # Step 6: Strip plurals
    cleaned = re.sub(r'(Disease|Disorder|Syndrome|Infection|Neoplasm|Fracture|Tumor|Ulcer)s\b',
                     r'\1', cleaned)
    
    # Step 7: British → American spelling
    cleaned = cleaned.replace('Leukaemia', 'Leukemia')
    cleaned = cleaned.replace('Tumour', 'Tumor')
    cleaned = cleaned.replace('Haemophilia', 'Hemophilia')
    cleaned = cleaned.replace('Anaemia', 'Anemia')
    cleaned = cleaned.replace('Oesophageal', 'Esophageal')
    
    return cleaned

# Test
tests = [
    "BREAST CANCER", "breast cancer", "Breast Cancer",
    "HIV INFECTIONS", "HIV Infections",
    "HEALTHY", "Healthy",
    "NSCLC Stage IV", "NSCLC",
    "COPD", "Copd",
    "Non-Small Cell Lung Cancer (NSCLC)",
    "ADHD - Attention Deficit Disorder With Hyperactivity",
    "Leukaemia (Acute Myeloid)",
    "Head and Neck Cancer", "HEAD AND NECK CANCER",
]
for t in tests:
    print(f"  '{t}' → '{clean_condition_v3(t)}'")
```

```python
# PURPOSE: Apply clean_condition_v3 to every row.
# WHY: Conditions are pipe-separated ("|"), same approach as country/drugs.
# Split by "|" → clean each → rejoin.

def clean_condition_cell(cell):
    """Clean a cell with multiple pipe-separated conditions."""
    if pd.isna(cell):
        return cell
    
    conditions = cell.split('|')
    cleaned = [clean_condition_v3(c.strip()) for c in conditions]
    
    # Remove duplicates keeping order
    seen = set()
    unique = []
    for c in cleaned:
        lower = c.lower()
        if lower not in seen:
            seen.add(lower)
            unique.append(c)
    
    return ' | '.join(unique)

# Apply
df['condition_cleaned'] = df['condition'].apply(clean_condition_cell)

# Count individual conditions before and after
before_conds = set()
for cell in df['condition'].dropna():
    for c in cell.split('|'):
        before_conds.add(c.strip())

after_conds = set()
for cell in df['condition_cleaned'].dropna():
    for c in cell.split('|'):
        after_conds.add(c.strip())

print(f"Unique individual conditions - Before: {len(before_conds)}")
print(f"Unique individual conditions - After:  {len(after_conds)}")
print(f"Reduced by: {len(before_conds) - len(after_conds)}")

print(f"\nUnique condition cells - Before: {df['condition'].nunique()}")
print(f"Unique condition cells - After:  {df['condition_cleaned'].nunique()}")
```

```python
# PURPOSE: Find near-duplicate conditions using fuzzy matching.
# WHY: "Diabetes Mellitus, Type 2" vs "Diabetes Mellitus Type 2" (just a comma)
# or "Lung Cancer" vs "Lung Cancers" that our plural rule missed.
# We use a HIGH threshold (95) because medical terms are precise —
# "Breast Cancer" and "Breast Caner" might be a typo but
# "Breast Cancer" and "Brain Cancer" are very different.

from collections import Counter

# Count frequency of each cleaned individual condition
cond_freq_after = Counter()
for cell in df['condition_cleaned'].dropna():
    for c in cell.split('|'):
        cond_freq_after[c.strip()] += 1

# Get unique conditions sorted by frequency (most common first)
unique_conds = sorted(cond_freq_after.keys(), key=lambda x: -cond_freq_after[x])

print(f"Unique conditions to compare: {len(unique_conds)}")
print("Running fuzzy matching (this may take a minute)...")

# Fuzzy match: compare each condition against more frequent ones
# Only check conditions with 2+ occurrences as targets (canonical candidates)
frequent_conds = [c for c in unique_conds if cond_freq_after[c] >= 2]
rare_conds = [c for c in unique_conds if cond_freq_after[c] == 1]

print(f"Frequent conditions (2+): {len(frequent_conds)}")
print(f"Rare conditions (1x): {len(rare_conds)}")

condition_map = {}
for rare in rare_conds:
    best_score = 0
    best_match = None
    for freq in frequent_conds:
        score = fuzz.ratio(rare.lower(), freq.lower())
        if score > best_score and score >= 95:
            best_score = score
            best_match = freq
    if best_match:
        condition_map[rare] = best_match

print(f"\nFuzzy matches found: {len(condition_map)}")
print(f"\nSample matches:")
for old, new in list(condition_map.items())[:20]:
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Remove bad fuzzy matches where stage/grade numbers differ.
# WHY: "Stage II" vs "Stage IV" are completely different medical stages.
# Fuzzy matching sees them as 95% similar but they're medically opposite.
# RULE: If both conditions contain "Stage" or "Grade", don't fuzzy match.

import re
notfuzzymatch = ['stage', 'grade', 'class', 'type']

def is_safe_fuzzy_match(rare, freq):
    """Check if this fuzzy match is safe (not a stage/grade mismatch)."""
    rare_lower = rare.lower()
    freq_lower = freq.lower()
    
    # If either condition contains a term that shouldn't be fuzzy matched, don't match
    if any(term in rare_lower for term in notfuzzymatch) or any(term in freq_lower for term in notfuzzymatch):
        # Extract term values
        for term in notfuzzymatch:
            rare_term = re.findall(rf'{term}\s*(\w+)', rare_lower)
            freq_term = re.findall(rf'{term}\s*(\w+)', freq_lower)
            if rare_term and freq_term and rare_term[0] != freq_term[0]:
                return False

    # If both have stage/grade numbers, don't match
    if ('stage' in rare_lower and 'stage' in freq_lower):
        # Extract stage values
        rare_stage = re.findall(r'stage\s*(\w+)', rare_lower)
        freq_stage = re.findall(r'stage\s*(\w+)', freq_lower)
        if rare_stage and freq_stage and rare_stage[0] != freq_stage[0]:
            return False
    
    if ('grade' in rare_lower and 'grade' in freq_lower):
        rare_grade = re.findall(r'grade\s*(\w+)', rare_lower)
        freq_grade = re.findall(r'grade\s*(\w+)', freq_lower)
        if rare_grade and freq_grade and rare_grade[0] != freq_grade[0]:
            return False
    
    # Don't match opposites
    opposite_pairs = [
        ('stable', 'unstable'),
        ('pre', 'post'),
        ('acute', 'chronic'),
        ('early', 'late'),
        ('primary', 'secondary'),
        ('benign', 'malignant'),
    ]
    for a, b in opposite_pairs:
        if (a in rare_lower and b in freq_lower) or (b in rare_lower and a in freq_lower):
            return False
    
    # Don't match different concepts that just sound similar
    if fuzz.token_sort_ratio(rare_lower, freq_lower) < 90:
        return False
    
    return True

# Re-run fuzzy matching with safety checks
condition_map_safe = {}
for rare in rare_conds:
    best_score = 0
    best_match = None
    for freq in frequent_conds:
        score = fuzz.ratio(rare.lower(), freq.lower())
        if score > best_score and score >= 95:
            if is_safe_fuzzy_match(rare, freq):
                best_score = score
                best_match = freq
    if best_match:
        condition_map_safe[rare] = best_match

print(f"Safe fuzzy matches: {len(condition_map_safe)}")
print(f"\nAll matches:")
for old, new in sorted(condition_map_safe.items()):
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Remove remaining bad fuzzy matches, then apply everything.

BAD_CONDITION_MATCHES = {
    'Prehabilitation',
    'B Acute Lymphoblastic Leukemia',
    'Metastatic Lung Small Cell Carcinoma',
    'Mental Health Wellness',
    'T-Prolymphocytic Leukemia',
    'Stage II Rectal Cancer Ajcc V7',
    'Stage III Rectal Cancer Ajcc V7',
    'Stage IV Colorectal Cancer Ajcc V7',
    'Prediabetes',   # → PreDiabetes is worse
    'Caregiver',     # singular is fine, don't map to plural
    'Drug Interaction',  # singular is fine
    'Mobile Application', # singular is fine
}

for bad in BAD_CONDITION_MATCHES:
    condition_map_safe.pop(bad, None)

print(f"Final safe matches: {len(condition_map_safe)}")

# Now apply: condition cleaning + fuzzy mapping
def clean_condition_cell_final(cell):
    """Full pipeline: clean each condition, then apply fuzzy map."""
    if pd.isna(cell):
        return cell
    
    conditions = cell.split('|')
    cleaned = []
    for c in conditions:
        name = clean_condition_v3(c.strip())
        # Apply fuzzy map
        if name in condition_map_safe:
            name = condition_map_safe[name]
        cleaned.append(name)
    
    # Remove duplicates keeping order
    seen = set()
    unique = []
    for c in cleaned:
        lower = c.lower()
        if lower not in seen:
            seen.add(lower)
            unique.append(c)
    
    return ' | '.join(unique)

# Apply
df['condition_cleaned'] = df['condition'].apply(clean_condition_cell_final)

# Count
before_conds = set()
for cell in df['condition'].dropna():
    for c in cell.split('|'):
        before_conds.add(c.strip())

after_conds = set()
for cell in df['condition_cleaned'].dropna():
    for c in cell.split('|'):
        after_conds.add(c.strip())

print(f"\nIndividual conditions - Before: {len(before_conds)}")
print(f"Individual conditions - After:  {len(after_conds)}")
print(f"Reduced by: {len(before_conds) - len(after_conds)}")

print(f"\nCondition cells - Before: {df['condition'].nunique()}")
print(f"Condition cells - After:  {df['condition_cleaned'].nunique()}")
```
