```python
# PURPOSE: Understand how pycountry works before using it.
# WHY: Like testing a new tool on one screw before using it on the whole shelf.
# pycountry is a library that has a list of all countries in the world.
# We give it a messy name → it gives back the official name.

import pycountry

# Test with a few examples
test_names = ["United States", "USA", "US", "Turkey (Türkiye)", "Korea, Republic of", "Brazil"]

for name in test_names:
    # Try exact match first
    result = pycountry.countries.get(name=name)
    if not result:
        # Try fuzzy search if exact fails
        try:
            result = pycountry.countries.search_fuzzy(name)[0]
        except LookupError:
            result = None
    
    clean = result.name if result else "NOT FOUND"
    print(f"  '{name}' → '{clean}'")
```

```python
# PURPOSE: Create a function that cleans ANY country name.
# WHY: pycountry works for most names, but some need manual help.
# Strategy: 1) Check manual fixes first → 2) Try pycountry exact → 3) Try fuzzy search
# Like a 3-step filter: easy fixes first, then smart lookup, then best guess.

import pycountry

# Manual fixes for names pycountry can't handle
COUNTRY_MANUAL_FIXES = {
    "turkey (türkiye)": "Turkey",
    "türkiye": "Turkey",
    "brasil": "Brazil",
    "korea, republic of": "South Korea",
    "republic of korea": "South Korea",
    "korea, democratic people's republic of": "North Korea",
    "russian federation": "Russia",
    "iran, islamic republic of": "Iran",
    "taiwan, province of china": "Taiwan",
    "viet nam": "Vietnam",
    "congo, the democratic republic of the": "Democratic Republic of the Congo",
    "tanzania, united republic of": "Tanzania",
    "palestinian territories": "Palestine",
    "faroe islands": "Faroe Islands",
}

def clean_country(name):
    """Takes a messy country name, returns a clean one."""
    if pd.isna(name):
        return name
    
    name_lower = name.strip().lower()
    
    # Step 1: Check manual fixes
    if name_lower in COUNTRY_MANUAL_FIXES:
        return COUNTRY_MANUAL_FIXES[name_lower]
    
    # Step 2: Try pycountry exact match
    result = pycountry.countries.get(name=name.strip())
    if result:
        return result.name
    
    # Step 3: Try fuzzy search
    try:
        result = pycountry.countries.search_fuzzy(name.strip())[0]
        return result.name
    except LookupError:
        return name.strip()  # give up, return as-is but trimmed

# Test it
test = ["Turkey (Türkiye)", "Brasil", "USA", "United States", "Korea, Republic of"]
for t in test:
    print(f"  '{t}' → '{clean_country(t)}'")
```

```python
# PURPOSE: Clean every country in the dataset.
# WHY: Remember, some rows have MULTIPLE countries like "France, Germany, Italy".
# So we: 1) Split by comma → 2) Clean each one → 3) Join back together.
# Like washing each sock individually then putting them back in the same drawer.

def clean_country_cell(cell):
    """Handles cells with multiple comma-separated countries."""
    if pd.isna(cell):
        return cell
    
    # Split by comma, clean each, rejoin
    countries = [clean_country(c.strip()) for c in cell.split(",")]
    # Remove duplicates while keeping order
    seen = set()
    unique = []
    for c in countries:
        if c not in seen:
            seen.add(c)
            unique.append(c)
    
    return ", ".join(unique)

# Apply to the whole column
df['country_cleaned'] = df['country'].apply(clean_country_cell)

# Check results
print(f"Before: {df['country'].nunique()} unique values")
print(f"After:  {df['country_cleaned'].nunique()} unique values")
```

```python
# PURPOSE: Some country names CONTAIN commas like "Korea, Republic of"
# or "Tanzania, United Republic of". Our comma-split broke them apart.
# FIX: Replace these tricky names BEFORE splitting by comma.
# Like replacing trick words before cutting a sentence into pieces.

# These are names that have commas inside them — we swap them first
COMMA_NAMES_FIX = {
    "Korea, Republic of": "South Korea",
    "Korea, Democratic People's Republic of": "North Korea",
    "Congo, The Democratic Republic of the": "Democratic Republic of the Congo",
    "Congo, the Democratic Republic of the": "Democratic Republic of the Congo",
    "Tanzania, United Republic of": "Tanzania",
    "Iran, Islamic Republic of": "Iran",
    "Taiwan, Province of China": "Taiwan",
    "Venezuela, Bolivarian Republic of": "Venezuela",
    "Bolivia, Plurinational State of": "Bolivia",
    "Moldova, Republic of": "Moldova",
    "Micronesia, Federated States of": "Micronesia",
    "Palestine, State of": "Palestine",
    "Lao People's Democratic Republic": "Laos",
    "Syrian Arab Republic": "Syria",
    "Viet Nam": "Vietnam",
    "Russian Federation": "Russia",
    "Czechia": "Czech Republic",
    "Côte d'Ivoire": "Ivory Coast",
    "Réunion": "Reunion",
}

def clean_country_cell_v2(cell):
    """Improved version: fixes comma-names BEFORE splitting."""
    if pd.isna(cell):
        return cell
    
    text = cell.strip()
    
    # Step 1: Replace known comma-containing names before splitting
    for messy, clean in COMMA_NAMES_FIX.items():
        text = text.replace(messy, clean)
    
    # Step 2: Now safe to split by comma
    countries = [clean_country(c.strip()) for c in text.split(",")]
    
    # Step 3: Remove duplicates keeping order
    seen = set()
    unique = []
    for c in countries:
        if c not in seen:
            seen.add(c)
            unique.append(c)
    
    return ", ".join(unique)

# Re-apply
df['country_cleaned'] = df['country'].apply(clean_country_cell_v2)

print(f"Before: {df['country'].nunique()} unique values")
print(f"After:  {df['country_cleaned'].nunique()} unique values")

# Check individual countries again
all_countries = df['country_cleaned'].dropna().str.split(", ").explode().str.strip()
print(f"\nUnique individual countries: {all_countries.nunique()}")
print(f"\nAll values:\n{sorted(all_countries.unique())}")
```

```python
# PURPOSE: Rebuild clean_country() with 3 layers:
# 1) PRE-FIX: handle known messy inputs before pycountry
# 2) PYCOUNTRY: try to match
# 3) POST-FIX: convert ugly official names to common names
# Like: wash → dry → iron. Three steps to clean clothes properly.

# POST-FIX: pycountry's ugly official names → common names
COUNTRY_POST_FIX = {
    "Iran, Islamic Republic of": "Iran",
    "Viet Nam": "Vietnam",
    "Russian Federation": "Russia",
    "Taiwan, Province of China": "Taiwan",
    "Tanzania, United Republic of": "Tanzania",
    "Syrian Arab Republic": "Syria",
    "Lao People's Democratic Republic": "Laos",
    "Venezuela, Bolivarian Republic of": "Venezuela",
    "Moldova, Republic of": "Moldova",
    "Réunion": "Reunion",
    "Korea, Republic of": "South Korea",
    "Korea, Democratic People's Republic of": "North Korea",
    "Congo, The Democratic Republic of the": "Democratic Republic of the Congo",
    "Bolivia, Plurinational State of": "Bolivia",
    "Micronesia, Federated States of": "Micronesia",
    "Palestine, State of": "Palestine",
    "Côte d'Ivoire": "Ivory Coast",
    "Czechia": "Czech Republic",
}

# PRE-FIX: handle names that pycountry can't find at all
COUNTRY_PRE_FIX = {
    "turkey (türkiye)": "Turkey",
    "türkiye": "Turkey",
    "palestinian territories": "Palestine",
    "faroe islands": "Faroe Islands",
    "brasil": "Brazil"
}

def clean_country_v3(name):
    """Final version: pre-fix → pycountry → post-fix."""
    if pd.isna(name):
        return name
    
    trimmed = name.strip()
    lower = trimmed.lower()
    
    # Step 1: Pre-fix
    if lower in COUNTRY_PRE_FIX:
        return COUNTRY_PRE_FIX[lower]
    
    # Step 2: Try pycountry
    result = pycountry.countries.get(name=trimmed)
    if not result:
        try:
            result = pycountry.countries.search_fuzzy(trimmed)[0]
        except LookupError:
            return trimmed
    
    official = result.name
    
    # Step 3: Post-fix
    if official in COUNTRY_POST_FIX:
        return COUNTRY_POST_FIX[official]
    
    return official

# Test
test = ["Iran", "Vietnam", "Russia", "Taiwan", "Tanzania",
        "Turkey (Türkiye)", "USA", "United States", "Brasil"]
for t in test:
    print(f"  '{t}' → '{clean_country_v3(t)}'")
```

```python
# PURPOSE: Clean sponsor names by stripping common suffixes and normalizing.
# WHY: "Novartis Pharmaceuticals" and "Novartis" are the same company.
# Strategy: 
#   1) Strip suffixes like Inc., LLC, Ltd, Co., Pharmaceuticals, etc.
#   2) Normalize whitespace and punctuation
#   3) Use fuzzy matching for near-duplicates
# Like peeling labels off jars to see they're the same brand.

import re

# Suffixes to strip (order matters — longer first)
SUFFIXES_TO_STRIP = [
    r",?\s*Inc\.?",
    r",?\s*LLC\.?",
    r",?\s*Ltd\.?",
    r",?\s*Co\.?",
    r",?\s*Corp\.?",
    r",?\s*Corporation",
    r",?\s*Incorporated",
    r",?\s*Limited",
    r",?\s*S\.?A\.?",
    r",?\s*S\.?p\.?A\.?",
    r",?\s*GmbH",
    r",?\s*AG",
    r",?\s*PLC",
    r",?\s*Pty",
    r",?\s*B\.?V\.?",
    r",?\s*N\.?V\.?",
    r",?\s*SE",
]

def clean_sponsor(name):
    """Clean a single sponsor name."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # Remove suffixes from the END of the name
    for suffix in SUFFIXES_TO_STRIP:
        cleaned = re.sub(suffix + r'\s*$', '', cleaned, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation
    cleaned = cleaned.rstrip('.,;: ')
    
    return cleaned

# Test
tests = ["Pfizer", "Pfizer, Inc.", "Pfizer Inc", 
         "Novartis Pharmaceuticals", "Novartis",
         "Click Therapeutics, Inc.", "Pharmavite LLC",
         "Hoffmann-La Roche", "Eli Lilly and Company"]
for t in tests:
    print(f"  '{t}' → '{clean_sponsor(t)}'")
```

```python
# PURPOSE: Extend suffix stripping to catch "Pharmaceuticals", "Vaccines", 
# "and Company" etc. These are department names, not different companies.
# Like: "John's Bakery Department" → just "John's Bakery" → just "John"

SUFFIXES_TO_STRIP_V2 = [
    # Corporate suffixes
    r",?\s*Inc\.?",
    r",?\s*LLC\.?",
    r",?\s*Ltd\.?",
    r",?\s*Co\.?",
    r",?\s*Corp\.?",
    r",?\s*Corporation",
    r",?\s*Incorporated",
    r",?\s*Limited",
    r",?\s*S\.?A\.?",
    r",?\s*S\.?p\.?A\.?",
    r",?\s*GmbH",
    r",?\s*PLC",
    r",?\s*Pty",
    r",?\s*B\.?V\.?",
    r",?\s*N\.?V\.?",
    r",?\s*SE",
    # Division/department suffixes
    r",?\s*Pharmaceuticals?",
    r",?\s*Vaccines?",
    r",?\s*and Company",
    r",?\s*& Company",
    r",?\s*Biosciences?",
    r",?\s*Therapeutics?",
    r",?\s*Sciences?",
    r",?\s*Biotech",
    r",?\s*Pharma",
]

def clean_sponsor_v2(name):
    """Clean sponsor: strip suffixes, normalize whitespace."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # Run suffix stripping TWICE (for "Novartis Pharmaceuticals, Inc.")
    for _ in range(2):
        for suffix in SUFFIXES_TO_STRIP_V2:
            cleaned = re.sub(suffix + r'\s*$', '', cleaned, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation
    cleaned = cleaned.rstrip('.,;: ')
    
    return cleaned

# Test
tests = ["Novartis Pharmaceuticals", "Novartis", "Novartis Vaccines",
         "Eli Lilly and Company", "Hoffmann-La Roche",
         "Click Therapeutics, Inc.", "Merck Sharp & Dohme LLC",
         "Boehringer Ingelheim", "AstraZeneca"]
for t in tests:
    print(f"  '{t}' → '{clean_sponsor_v2(t)}'")
```

```python
# PURPOSE: Fix over-stripping. "Click Therapeutics" is the real name,
# but "Novartis Pharmaceuticals" is Novartis + department label.
# NEW STRATEGY: 
#   1) Strip corporate suffixes (Inc, LLC, Ltd, etc.) — always safe
#   2) Strip division words (Pharmaceuticals, Vaccines) ONLY for known big pharma
# Like: remove "Mr." from everyone, but only remove "Junior" from known families.

import re

CORPORATE_SUFFIXES = [
    r",?\s*Inc\.?",
    r",?\s*LLC\.?",
    r",?\s*Ltd\.?",
    r",?\s*Co\.?\b",
    r",?\s*Corp\.?",
    r",?\s*Corporation",
    r",?\s*Incorporated",
    r",?\s*Limited",
    r",?\s*S\.?A\.?\b",
    r",?\s*S\.?p\.?A\.?",
    r",?\s*GmbH",
    r",?\s*PLC",
    r",?\s*Pty",
    r",?\s*B\.?V\.?\b",
    r",?\s*N\.?V\.?\b",
    r",?\s*SE\b",
    r",?\s*and Company",
    r",?\s*& Company",
]

# Known big pharma where "Pharmaceuticals"/"Vaccines" is a division label
KNOWN_PHARMA_BASES = [
    "novartis", "pfizer", "roche", "merck", "sanofi", "bayer",
    "johnson", "abbvie", "amgen", "gilead", "bristol-myers",
    "astrazeneca", "glaxosmithkline", "gsk", "takeda", "allergan",
]

def clean_sponsor_v3(name):
    """Clean sponsor with safe suffix stripping."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # Step 1: Strip corporate suffixes (always safe)
    for suffix in CORPORATE_SUFFIXES:
        cleaned = re.sub(suffix + r'\s*$', '', cleaned, flags=re.IGNORECASE).strip()
    
    # Step 2: Strip division words ONLY if base is a known big pharma
    base_lower = cleaned.lower().split()[0] if cleaned else ""
    if any(pharma in cleaned.lower() for pharma in KNOWN_PHARMA_BASES):
        for div in [r"\s+Pharmaceuticals?$", r"\s+Vaccines?$", r"\s+Pharma$", 
                     r"\s+Biosciences?$", r"\s+Biotech$"]:
            cleaned = re.sub(div, '', cleaned, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation
    cleaned = cleaned.rstrip('.,;: ')
    
    return cleaned

# Test
tests = ["Novartis Pharmaceuticals", "Novartis Vaccines", "Novartis",
         "Click Therapeutics, Inc.", "Eli Lilly and Company",
         "Merck Sharp & Dohme LLC", "Hoffmann-La Roche",
         "Pfizer's Upjohn has merged with Mylan to form Viatris Inc.",
         "Metsera, a wholly owned subsidiary of Pfizer"]
for t in tests:
    print(f"  '{t}' → '{clean_sponsor_v3(t)}'")
```

```python
# PURPOSE: Handle weird edge cases with a manual map, then apply to all 5000 rows.
# WHY: Some sponsor names are sentences or contain subsidiary notes.
# These are too rare and weird for rules — just map them by hand.
# Like: most clothes go in the washing machine, but silk needs hand wash.

SPONSOR_MANUAL_MAP = {
    "Pfizer's Upjohn has merged with Mylan to form Viatris Inc.": "Viatris",
    "Metsera, a wholly owned subsidiary of Pfizer": "Pfizer",
    "Wyeth is now a wholly owned subsidiary of Pfizer": "Pfizer",
}

def clean_sponsor_final(name):
    """Final sponsor cleaner: manual map → suffix stripping."""
    if pd.isna(name):
        return name
    
    trimmed = name.strip()
    
    # Step 1: Check manual map first
    if trimmed in SPONSOR_MANUAL_MAP:
        return SPONSOR_MANUAL_MAP[trimmed]
    
    # Step 2: Apply suffix stripping
    return clean_sponsor_v3(trimmed)

# Apply to whole dataset
df['sponsor_cleaned'] = df['sponsor'].apply(clean_sponsor_final)

print(f"Before: {df['sponsor'].nunique()} unique sponsors")
print(f"After:  {df['sponsor_cleaned'].nunique()} unique sponsors")
print(f"Reduced by: {df['sponsor'].nunique() - df['sponsor_cleaned'].nunique()}")
```

```python
# PURPOSE: Find what types of organizations exist in our sponsor names.
# WHY: Before making rules, we need to see the real patterns.
# We'll extract the LAST word(s) from each sponsor to find common suffixes.
# Like sorting a pile of business cards by company type.

from collections import Counter

# Get all unique sponsors
sponsors = df['sponsor'].unique()

# Extract last 1-2 words as potential suffix
last_words = []
for s in sponsors:
    words = s.strip().rstrip('.,;').split()
    if len(words) >= 2:
        last_words.append(words[-1])
        last_words.append(words[-2] + " " + words[-1])

# Count most common endings
suffix_counts = Counter(last_words)

print("=== TOP 200 MOST COMMON ENDINGS ===")
for suffix, count in suffix_counts.most_common(200):
    if count >= 2:  # only show if appears 3+ times
        print(f"  {count:4d}x  '{suffix}'")
```

```python
# PURPOSE: Standardize all corporate suffix variations to ONE canonical form.
# WHY: "Inc.", "Inc", "Incorporated" are the same thing written differently.
# We pick one spelling and force all to match.
# Like making everyone write dates as DD/MM/YYYY instead of mixing formats.

import re

def standardize_sponsor_v3(name):
    """Standardize sponsor suffix variations to canonical forms."""
    if pd.isna(name):
        return name
    
    cleaned = name.strip()
    
    # 1. Standardize "Co., Ltd" variations → "Co Ltd"
    #    Handles: Co., Ltd / Co. Ltd / Co.,Ltd / Co.,Ltd. / Co., Ltd.
    cleaned = re.sub(r'\bCo\.\s*,?\s*Ltd\.?', 'Co Ltd', cleaned)
    
    # 2. Standardize Incorporated → Inc
    cleaned = re.sub(r',?\s*Incorporated\s*$', ' Inc', cleaned)
    
    # 3. Standardize "Inc." or ", Inc." or ", Inc" → " Inc"
    cleaned = re.sub(r',?\s*Inc\.?\s*$', ' Inc', cleaned)
    
    # 4. Standardize Limited / LTD → Ltd
    cleaned = re.sub(r',?\s*Limited\s*$', ' Ltd', cleaned)
    cleaned = re.sub(r',?\s*LTD\.?\s*$', ' Ltd', cleaned)
    cleaned = re.sub(r',?\s*Ltd\.?\s*$', ' Ltd', cleaned)
    
    # 5. Standardize Corporation / Corp. → Corp
    cleaned = re.sub(r',?\s*Corporation\s*$', ' Corp', cleaned)
    cleaned = re.sub(r',?\s*Corp\.?\s*$', ' Corp', cleaned)
    
    # 6. Standardize LLC. → LLC
    cleaned = re.sub(r',?\s*LLC\.?\s*$', ' LLC', cleaned)
    
    # 7. Standardize Pharmaceuticals / Pharmaceutical → Pharma
    #cleaned = re.sub(r',?\s*Pharmaceuticals?\s*$', ' Pharma', cleaned)
    
    # 8. Standardize Pty Ltd → Ltd
    cleaned = re.sub(r',?\s*Pty\s+Ltd\.?\s*$', ' Ltd', cleaned)
    
    # 9. Standardize gGmbH → GmbH
    cleaned = re.sub(r'\bgGmbH\s*$', 'GmbH', cleaned)
    
    # 10. Standardize S.p.A. variations
    cleaned = re.sub(r',?\s*S\.?p\.?A\.?\s*$', ' SpA', cleaned)
    
    # 11. Standardize B.V. → BV
    cleaned = re.sub(r',?\s*B\.?V\.?\s*$', ' BV', cleaned)
    
    # 12. Standardize N.V. → NV
    cleaned = re.sub(r',?\s*N\.?V\.?\s*$', ' NV', cleaned)
    
    # Clean up extra spaces
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    return cleaned

# Test
tests = [
    "Milestone Pharmaceuticals Inc.",
    "Novartis Pharmaceuticals",
    "Click Therapeutics, Inc.",
    "Celltex Therapeutics Corporation",
    "SK Chemicals Co., Ltd.",
    "Eli Lilly and Company",
    "Merck Sharp & Dohme LLC",
    "Staidson Co.,Ltd",
]
for t in tests:
    print(f"  '{t}' → '{standardize_sponsor_v3(t)}'")
```

```python
# PURPOSE: Handle weird edge cases manually, then apply to all 5000 rows.
# WHY: Some sponsors are sentences, not names. Too weird for rules.

SPONSOR_MANUAL_MAP = {
    "Pfizer's Upjohn has merged with Mylan to form Viatris Inc.": "Viatris Inc",
    "Metsera, a wholly owned subsidiary of Pfizer": "Pfizer",
    "Wyeth is now a wholly owned subsidiary of Pfizer": "Pfizer",
}

def clean_sponsor_final(name):
    """Manual map first, then standardize suffixes."""
    if pd.isna(name):
        return name
    
    trimmed = name.strip()
    
    if trimmed in SPONSOR_MANUAL_MAP:
        return SPONSOR_MANUAL_MAP[trimmed]
    
    return standardize_sponsor_v3(trimmed)

# Apply
df['sponsor_cleaned'] = df['sponsor'].apply(clean_sponsor_final)

print(f"Before: {df['sponsor'].nunique()} unique")
print(f"After:  {df['sponsor_cleaned'].nunique()} unique")
print(f"Reduced by: {df['sponsor'].nunique() - df['sponsor_cleaned'].nunique()}")
```

```python
# PURPOSE: Extract the "core" of a sponsor name by stripping ALL suffixes.
# WHY: We only use this core for COMPARING names, not as the final output.
# The final output keeps the standardized suffix.
# Like comparing two people by face (core) not by their hat (suffix).

SUFFIXES_TO_STRIP_FOR_CORE = [
    r'\bCo\s*Ltd$',
    r'\bInc$',
    r'\bLtd$',
    r'\bLLC$',
    r'\bCorp$',
    r'\bGmbH$',
    r'\bSpA$',
    r'\bBV$',
    r'\bNV$',
    r'\bSA$',
    r'\bSL$',
    r'\bSAS$',
    r'\bAG$',
    r'\bAB$',
    r'\bA/S$',
    r'\bPLC$',
    r'\bSE$',
    r'\bPLLC$',
    r'\band Company$',
    r'\b& Company$',
]

def get_core_name(name):
    """Strip all suffixes to get the core name for comparison only."""
    if pd.isna(name):
        return name
    
    core = name.strip()
    
    for suffix in SUFFIXES_TO_STRIP_FOR_CORE:
        core = re.sub(suffix, '', core, flags=re.IGNORECASE).strip()
    
    # Remove trailing punctuation and extra spaces
    core = core.rstrip('.,;: ')
    core = re.sub(r'\s+', ' ', core).strip()
    
    return core

# Test: compare core names
test_pairs = [
    ("Pfizer Inc", "Pfizer"),
    ("Novartis Pharmaceuticals", "Novartis"),
    ("Cairo University", "Cairo University Hospital"),
    ("Eli Lilly and Company", "Eli Lilly"),
    ("Merck Sharp & Dohme LLC", "Merck Sharp & Dohme"),
    ("University of California, San Francisco", "University of California San Francisco"),
]

for a, b in test_pairs:
    core_a = get_core_name(a)
    core_b = get_core_name(b)
    score = fuzz.token_sort_ratio(core_a, core_b)
    print(f"  '{a}' → core: '{core_a}'")
    print(f"  '{b}' → core: '{core_b}'")
    print(f"    score: {score}")
    print()
```

```python
# PURPOSE: Add division words like "Pharmaceuticals", "Therapeutics" etc.
# to the core stripper FOR COMPARISON ONLY.
# WHY: "Novartis Pharmaceuticals" and "Novartis" are the same company.
# We strip "Pharmaceuticals" only when comparing, NOT in the final output.
# Like ignoring someone's job title when checking if they're the same person.

SUFFIXES_TO_STRIP_FOR_CORE = [
    # Corporate legal suffixes
    r'\bCo\s*Ltd$',
    r'\bInc$',
    r'\bLtd$',
    r'\bLLC$',
    r'\bCorp$',
    r'\bGmbH$',
    r'\bSpA$',
    r'\bBV$',
    r'\bNV$',
    r'\bSA$',
    r'\bSL$',
    r'\bSAS$',
    r'\bAG$',
    r'\bAB$',
    r'\bA/S$',
    r'\bPLC$',
    r'\bSE$',
    r'\bPLLC$',
    r'\band Company$',
    r'\b& Company$',
    # Division words
    r'\bPharmaceuticals?$',
    r'\bPharma$',
    r'\bVaccines?$',
    r'\bBiosciences?$',
    r'\bBiotech$',
    r'\bTherapeutics?$',
    r'\bBiopharma$',
    r'\bBiopharmaceuticals?$',
]

def get_core_name(name):
    """Strip all suffixes to get core name for comparison only."""
    if pd.isna(name):
        return name
    
    core = name.strip()
    
    # Run twice to catch combos like "Therapeutics Corp"
    for _ in range(2):
        for suffix in SUFFIXES_TO_STRIP_FOR_CORE:
            core = re.sub(suffix, '', core, flags=re.IGNORECASE).strip()
    
    core = core.rstrip('.,;: ')
    core = re.sub(r'\s+', ' ', core).strip()
    
    return core

# Test again
test_pairs = [
    ("Novartis Pharmaceuticals", "Novartis"),
    ("Novartis Vaccines", "Novartis"),
    ("Novartis Vaccines", "Novartis Pharmaceuticals"),
    ("Cairo University", "Cairo University Hospital"),
    ("Celltex Therapeutics Corp", "Celltex"),
    ("Samyang Biopharmaceuticals Corp", "Samyang"),
]

for a, b in test_pairs:
    core_a = get_core_name(a)
    core_b = get_core_name(b)
    score = fuzz.token_sort_ratio(core_a, core_b)
    print(f"  '{a}' → core: '{core_a}'")
    print(f"  '{b}' → core: '{core_b}'")
    print(f"    score: {score}")
    print()
```

```python
# PURPOSE: Group sponsors by their core name, then pick the most frequent
# version as the canonical name. Map all variations to it.
# WHY: "Novartis Vaccines" and "Novartis Pharmaceuticals" share core "Novartis"
# → they're the same company → map both to the most common version.
# Like grouping all mail for the Smith family into one mailbox.

from collections import defaultdict

# Step 1: Get all unique cleaned sponsors and their core names
unique_sponsors = df['sponsor_cleaned'].unique()
sponsor_cores = {s: get_core_name(s).lower() for s in unique_sponsors}

# Step 2: Group by exact core match
core_groups = defaultdict(list)
for sponsor, core in sponsor_cores.items():
    core_groups[core].append(sponsor)

# Step 3: Pick most frequent as canonical, build mapping
sponsor_map = {}
groups_found = 0

for core, members in core_groups.items():
    if len(members) > 1:
        groups_found += 1
        # Count how often each version appears in data
        freqs = {m: (df['sponsor_cleaned'] == m).sum() for m in members}
        canonical = max(freqs, key=freqs.get)
        for m in members:
            if m != canonical:
                sponsor_map[m] = canonical

print(f"Groups with multiple versions: {groups_found}")
print(f"Sponsors to remap: {len(sponsor_map)}")
print(f"\nAll mappings:")
for old, new in sorted(sponsor_map.items()):
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Fuzzy match core names to find more groups, then pick shortest
# name as canonical with the most common suffix attached.
# WHY: Exact core matching only found 5 groups. Fuzzy matching catches
# near-duplicates like slight spelling differences.
# Like sorting mail: even if "Jon Smith" and "John Smith" are slightly 
# different, they probably live in the same house.

from collections import defaultdict
from thefuzz import fuzz

# Step 1: Get unique cores
unique_sponsors = df['sponsor_cleaned'].unique()
sponsor_cores = {s: get_core_name(s) for s in unique_sponsors}

# Get unique cores and their members
core_to_sponsors = defaultdict(list)
for sponsor, core in sponsor_cores.items():
    core_to_sponsors[core.lower()].append(sponsor)

# Step 2: Fuzzy match between cores (threshold 90)
unique_cores = list(core_to_sponsors.keys())
merged_cores = {}  # maps core → group leader core

for i, core_a in enumerate(unique_cores):
    if core_a in merged_cores:
        continue
    for j in range(i + 1, len(unique_cores)):
        core_b = unique_cores[j]
        if core_b in merged_cores:
            continue
        score = fuzz.token_sort_ratio(core_a, core_b)
        if score >= 90:
            merged_cores[core_b] = core_a

# Step 3: Build final groups
final_groups = defaultdict(list)
for core, sponsors in core_to_sponsors.items():
    leader = merged_cores.get(core, core)
    final_groups[leader].extend(sponsors)

# Step 4: Pick shortest name as canonical
sponsor_map = {}
groups_found = 0

for leader, members in final_groups.items():
    if len(members) > 1:
        groups_found += 1
        canonical = min(members, key=len)
        for m in members:
            if m != canonical:
                sponsor_map[m] = canonical

print(f"Groups with multiple versions: {groups_found}")
print(f"Sponsors to remap: {len(sponsor_map)}")
print(f"\nAll mappings:")
for old, new in sorted(sponsor_map.items()):
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Fuzzy match ONLY pharma/biotech companies, not universities/hospitals.
# WHY: Universities all look similar ("University of X") causing false merges.
# Pharma companies have unique names so fuzzy matching is safe.
# Like: you can sort candy by color, but don't sort identical marbles by color.

from collections import defaultdict
from thefuzz import fuzz

# Keywords that indicate an institution (skip fuzzy for these)
INSTITUTION_KEYWORDS = [
    'university', 'hospital', 'clinic', 'institute', 'college',
    'medical center', 'cancer center', 'health system', 'school',
    'foundation', 'ministry', 'department', 'va ', 'veterans',
    'national', 'centre', 'center', 'nhs',
]

def is_institution(name):
    """Check if sponsor is a university/hospital/institution."""
    lower = name.lower()
    return any(kw in lower for kw in INSTITUTION_KEYWORDS)

# Step 1: Get cores and split into pharma vs institution
unique_sponsors = df['sponsor_cleaned'].unique()
sponsor_cores = {s: get_core_name(s) for s in unique_sponsors}

core_to_sponsors = defaultdict(list)
for sponsor, core in sponsor_cores.items():
    core_to_sponsors[core.lower()].append(sponsor)

# Step 2: Separate pharma cores from institution cores
pharma_cores = {}
institution_cores = {}
for core, sponsors in core_to_sponsors.items():
    if any(is_institution(s) for s in sponsors):
        institution_cores[core] = sponsors
    else:
        pharma_cores[core] = sponsors

print(f"Pharma/biotech cores: {len(pharma_cores)}")
print(f"Institution cores: {len(institution_cores)}")

# Step 3: Fuzzy match ONLY pharma cores (threshold 90)
pharma_core_list = list(pharma_cores.keys())
merged_cores = {}

for i, core_a in enumerate(pharma_core_list):
    if core_a in merged_cores:
        continue
    for j in range(i + 1, len(pharma_core_list)):
        core_b = pharma_core_list[j]
        if core_b in merged_cores:
            continue
        score = fuzz.token_sort_ratio(core_a, core_b)
        if score >= 90:
            merged_cores[core_b] = core_a

# Step 4: Build final groups (pharma fuzzy + institution exact)
final_groups = defaultdict(list)

# Pharma: use fuzzy groups
for core, sponsors in pharma_cores.items():
    leader = merged_cores.get(core, core)
    final_groups[leader].extend(sponsors)

# Institution: exact core only
for core, sponsors in institution_cores.items():
    final_groups[core].extend(sponsors)

# Step 5: Pick shortest name as canonical
sponsor_map = {}
groups_found = 0

for leader, members in final_groups.items():
    if len(members) > 1:
        groups_found += 1
        canonical = min(members, key=len)
        for m in members:
            if m != canonical:
                sponsor_map[m] = canonical

print(f"\nGroups with multiple versions: {groups_found}")
print(f"Sponsors to remap: {len(sponsor_map)}")
print(f"\nAll mappings:")
for old, new in sorted(sponsor_map.items()):
    print(f"  '{old}' → '{new}'")
```

```python
# PURPOSE: Apply the sponsor mapping to the dataset and see final results.
# WHY: We've verified all 8 mappings look correct. Time to apply.

# Apply: first standardize suffixes, then remap groups
def clean_sponsor_complete(name):
    """Full pipeline: manual map → standardize suffix → group remap."""
    if pd.isna(name):
        return name
    
    trimmed = name.strip()
    
    # Step 1: Manual map (weird sentences)
    if trimmed in SPONSOR_MANUAL_MAP:
        trimmed = SPONSOR_MANUAL_MAP[trimmed]
    
    # Step 2: Standardize suffixes
    cleaned = standardize_sponsor_v3(trimmed)
    
    # Step 3: Group remap
    if cleaned in sponsor_map:
        cleaned = sponsor_map[cleaned]
    
    return cleaned

df['sponsor_cleaned'] = df['sponsor'].apply(clean_sponsor_complete)

before = df['sponsor'].nunique()
after = df['sponsor_cleaned'].nunique()
print(f"Before: {before} unique sponsors")
print(f"After:  {after} unique sponsors")
print(f"Reduced by: {before - after}")
```

```python
# PURPOSE: Build a function to clean individual drug names.
# WHAT IT DOES:
#   1) Strip dosage info (numbers + mg/mcg/ml/%)
#   2) Extract aliases from parentheses (Chinese or regular)
#   3) Normalize casing to Title Case
#   4) Clean whitespace
# WHY: Get names to a consistent base form before API lookups.
# Like removing packaging before comparing products.

import re

def clean_drug_name(name):
    """Clean a single drug name (not the whole cell, just one drug)."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Remove dosage patterns like "10 mg", "500mg", "1.5 mcg/kg", "0.9%"
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    
    # Step 2: Remove "daily", "weekly", "twice daily" etc.
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 3: Remove "in X ML Injection" type patterns
    cleaned = re.sub(r'\bin\s+\d+\s*ML\s+\w+', '', cleaned, flags=re.IGNORECASE)
    
    # Step 4: Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    # Step 5: Title case (but preserve ALL-CAPS abbreviations like "CAR" "BCMA")
    if cleaned and not cleaned.isupper():
        cleaned = cleaned.title()
    
    # Step 6: Remove trailing/leading punctuation
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "Henagliflozin 10 mg daily",
    "Aspirin 81mg",
    "CR845 0.5 mcg/kg",
    "Bridion 200 MG in 2 ML Injection",
    "Pembrolizumab",
    "pembrolizumab",
    "KEYTRUDA",
    "Placebo",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name(t)}'")
```

```python
# PURPOSE: Fix two issues:
#   1) "Bridion In Injection" — remove orphaned words after dosage stripping
#   2) "KEYTRUDA" — title case ALL-CAPS names too (they're brand names)
# WHY: Dosage removal leaves behind words like "in", "Injection", "Capsule".
# And brand names in ALL CAPS should become Title Case for consistency.

def clean_drug_name_v2(name):
    """Improved drug name cleaner."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Remove dosage patterns
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    
    # Step 2: Remove pharmaceutical form words (leftover after dosage removal)
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    
    # Step 3: Remove time/frequency words
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 4: Remove orphaned small words ("in", "of", "for") at boundaries
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 5: Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    # Step 6: Title Case everything
    if cleaned:
        cleaned = cleaned.title()
    
    # Step 7: Remove trailing/leading punctuation
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "Bridion 200 MG in 2 ML Injection",
    "KEYTRUDA",
    "pembrolizumab",
    "Henagliflozin 10 mg daily",
    "Aspirin 81mg",
    "Placebo",
    "Sodium Oligomannate Capsules",
    "CR845 0.5 mcg/kg",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v2(t)}'")
```

```python
# PURPOSE: Fix Title Case breaking research codes like "CR845" → "Cr845".
# WHY: Research codes are alphanumeric (letters + numbers) like "CR845", 
# "ABT-089", "MK-3475". Title Case ruins them.
# RULE: If a word has letters AND numbers mixed, keep it UPPERCASE.
# Like: don't correct a license plate's spelling.

def smart_title_case(text):
    """Title case but preserve research codes (letters + numbers mixed)."""
    words = text.split()
    result = []
    for word in words:
        # If word mixes letters and digits → keep uppercase
        has_letter = any(c.isalpha() for c in word)
        has_digit = any(c.isdigit() for c in word)
        if has_letter and has_digit:
            result.append(word.upper())
        else:
            result.append(word.title())
    return ' '.join(result)

def clean_drug_name_v3(name):
    """Drug name cleaner with smart casing."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    # Step 1: Remove dosage patterns
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    
    # Step 2: Remove pharmaceutical form words
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    
    # Step 3: Remove time/frequency words
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 4: Remove orphaned small words
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    
    # Step 5: Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    # Step 6: Smart Title Case
    if cleaned:
        cleaned = smart_title_case(cleaned)
    
    # Step 7: Remove trailing/leading punctuation
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "CR845 0.5 mcg/kg",
    "ABT-089",
    "MK-3475",
    "TAS-117",
    "KEYTRUDA",
    "pembrolizumab",
    "Bridion 200 MG in 2 ML Injection",
    "Sodium Oligomannate Capsules",
    "5-Amino Levulinic Acid",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v3(t)}'")
```

```python
# PURPOSE: Fix "5-Amino" being treated as a research code.
# WHY: "5-Amino" has digit + letters but it's NOT a code — it's a chemical prefix.
# RULE: Only keep uppercase if digits and letters are MIXED together
# like "CR845", "ABT089". Not if separated by hyphen like "5-Amino".
# Better check: the word without hyphens still mixes letters+digits in same chunk.

def smart_title_case_v2(text):
    """Title case but preserve true research codes."""
    words = text.split()
    result = []
    for word in words:
        # Split by hyphen and check each part
        parts = word.split('-')
        is_code = False
        for part in parts:
            has_letter = any(c.isalpha() for c in part)
            has_digit = any(c.isdigit() for c in part)
            if has_letter and has_digit:
                is_code = True
                break
        
        if is_code:
            result.append(word.upper())
        else:
            result.append(word.title())
    return ' '.join(result)

def clean_drug_name_v4(name):
    """Drug name cleaner with improved casing."""
    if pd.isna(name) or not name.strip():
        return name
    
    cleaned = name.strip()
    
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    if cleaned:
        cleaned = smart_title_case_v2(cleaned)
    
    cleaned = cleaned.strip('.,;:- ')
    
    return cleaned if cleaned else name.strip()

# Test
tests = [
    "5-Amino Levulinic Acid",
    "CR845 0.5 mcg/kg",
    "ABT-089",
    "MK-3475",
    "KEYTRUDA",
    "pembrolizumab",
    "Bridion 200 MG in 2 ML Injection",
    "IL-12",
    "SHR-1210",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v4(t)}'")
```

```python
# PURPOSE: Better rule for research codes.
# INSIGHT: Research codes follow a pattern: 2-4 UPPERCASE LETTERS + hyphen + DIGITS
# like ABT-089, MK-3475, SHR-1210, IL-12, CR845.
# If the ORIGINAL word was ALL UPPERCASE and short, it's likely a code.
# RULE: If the original word is ALL CAPS (or all caps + digits), keep uppercase.

def smart_title_case_v3(text, original):
    """Title case but preserve research codes based on original casing."""
    words = text.split()
    orig_words = original.split()
    
    result = []
    for i, word in enumerate(words):
        # Find matching original word if possible
        orig = orig_words[i] if i < len(orig_words) else word
        
        # Rule 1: If original was ALL UPPERCASE → keep uppercase
        # Catches: ABT-089, MK-3475, KEYTRUDA, IL-12, SHR-1210, CR845
        if orig.replace('-', '').replace('.', '').isupper() and any(c.isalpha() for c in orig):
            result.append(word.upper())
        else:
            result.append(word.title())
    
    return ' '.join(result)

def clean_drug_name_v5(name):
    """Drug name cleaner with proper code detection."""
    if pd.isna(name) or not name.strip():
        return name
    
    original = name.strip()
    cleaned = original
    
    cleaned = re.sub(r'\d+\.?\d*\s*(?:mg|mcg|ug|ml|g\b|%|MG|ML|mcg/kg)\S*', '', cleaned)
    cleaned = re.sub(r'\b(?:Injection|Capsule|Capsules|Tablet|Tablets|Solution|Cream|Gel|Ointment|Syrup|Suspension|Powder)\b', 
                     '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:daily|weekly|once|twice|per day|per week)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\b(?:in|of|for|the|a|an)\b', '', cleaned, flags=re.IGNORECASE)
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    cleaned = cleaned.strip('.,;:- ')
    
    if not cleaned:
        return original
    
    # Smart casing: compare each word to original
    cleaned = smart_title_case_v3(cleaned, original)
    
    return cleaned

# Test
tests = [
    "ABT-089", "MK-3475", "SHR-1210", "IL-12", "CR845",
    "KEYTRUDA", "pembrolizumab", "5-Amino Levulinic Acid",
    "Bridion 200 MG in 2 ML Injection",
]
for t in tests:
    print(f"  '{t}' → '{clean_drug_name_v5(t)}'")
```

```python
# PURPOSE: Apply clean_drug_name_v5 to every row.
# WHY: Remember, many cells have multiple drugs separated by "+".
# So we: split by "+" → clean each drug → rejoin.
# Same approach as country (split, clean, rejoin).

def clean_drug_cell(cell):
    """Clean a cell that may contain multiple drugs separated by +."""
    if pd.isna(cell):
        return cell
    
    # Split by "+"
    drugs = cell.split('+')
    
    # Clean each drug individually
    cleaned = [clean_drug_name_v5(d.strip()) for d in drugs]
    
    # Remove empty strings and rejoin
    cleaned = [d for d in cleaned if d.strip()]
    
    return ' + '.join(cleaned)

# Apply to all rows
df['intervention_cleaned'] = df['intervention_name'].apply(clean_drug_cell)

print(f"Before: {df['intervention_name'].nunique()} unique")
print(f"After:  {df['intervention_cleaned'].nunique()} unique")
print(f"Reduced by: {df['intervention_name'].nunique() - df['intervention_cleaned'].nunique()}")

# Sample changes
changed = df[df['intervention_name'] != df['intervention_cleaned']][['intervention_name', 'intervention_cleaned']]
print(f"\nRows changed: {len(changed)}")
print(f"\nSample changes:")
for _, row in changed.sample(10, random_state=42).iterrows():
    print(f"  '{row['intervention_name']}'")
    print(f"  → '{row['intervention_cleaned']}'")
    print()
```
